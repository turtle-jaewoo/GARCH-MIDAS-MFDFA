### Data Preparation Function

```python
def download_btc_data(start_year, end_year, timeframe, symbol = 'BTC/USDT'):
    """
    Args:
        start_year (int): 시작 연도
        end_year (int): 종료 연도
        timeframe (str): 데이터 간격 ('1m', '10m' ,'1d')
        symbol (str): 코인 심볼 (기본 BTC/USDT)
    """

    save_dir = "data"
    if not os.path.exists(save_dir):
        os.makedirs(save_dir)
        print(f"[INFO] Directory '{save_dir}' created")

    # Create Binance object
    binance = ccxt.binance({
        'enableRateLimit': True,
        'options': {'defaultType': 'future'}
    })

    # Check wheather Binance support timeframe
    if timeframe not in binance.timeframes:
        print(f"[ERROR] '{timeframe} is not a valid timeframe")
        return

    print(f"Target: {symbol} | Timeframe: {timeframe} | Period: {start_year} ~ {end_year}")

    safe_symbol = symbol.replace("/", "")

    for year in range(start_year, end_year + 1):
        # Set file path including data folder
        filename = f"{safe_symbol}_{timeframe}_{year}.csv"
        file_path = os.path.join(save_dir, filename)

        # Check file existence
        if os.path.exists(file_path):
            print(f"[SKIP] '{filename}' already exists")
            continue

        print(f"\n[INFO] Downloading Year: {year}")
        start_date = datetime(year, 1, 1, 0, 0, 0, tzinfo = timezone.utc)
        end_date = datetime(year + 1, 1, 1, 0, 0, 0, tzinfo = timezone.utc)

        since = int(start_date.timestamp() * 1000)
        end_timestamp = int(end_date.timestamp() * 1000)

        year_ohlcvs = []

        approx_counts = {
            '1m': 525600, '1d': 365 # 10m Data Not Support in Binance API
        }
        total_candles = approx_counts.get(timeframe, 10000)

        pbar = tqdm(total = total_candles, desc = f"{year} Progress", unit = "candle")

        while since < end_timestamp:
            try:
                ohlcvs = binance.fetch_ohlcv(symbol, timeframe = timeframe, since = since, limit = 1000)

                if not ohlcvs:
                    break

                year_ohlcvs.extend(ohlcvs)

                current_last_time = ohlcvs[-1][0]
                since = current_last_time + 1

                pbar.update(len(ohlcvs))

                if current_last_time >= end_timestamp:
                    break

                time.sleep(0.05)

            except Exception as e:
                print(f"[ERROR] {e}, Retryung")
                time.sleep(3)
                continue

        pbar.close()

        if year_ohlcvs:
            df = pd.DataFrame(year_ohlcvs, columns = ['timestamp', 'open', 'high', 'low', 'close', 'volume'])
            df['datetime'] = pd.to_datetime(df['timestamp'], unit = 'ms', utc = True)

            df = df[(df['datetime'] >= start_date) & (df['datetime'] < end_date)]

            df.set_index('datetime', inplace = True)
            df.drop(columns = ['timestamp'], inplace = True)

            df.to_csv(file_path)
            print(f"[SUCCESS] Saved {file_path} (Rows: {len(df)})")
        else:
            print(f"[FAIL] No data for {year}")


def resample_to_10m(start_year, end_year, symbol = 'BTC/USDT'):
    """
    Transform 1m data to 10m data and store to csv file
    """
    save_dir = "data"
    safe_symbol = symbol.replace("/", "")

    print(f"Starting Resampling: {symbol} (1m -> 10m) | {start_year} ~ {end_year}")

    for year in range(start_year, end_year + 1):
        input_filename = f"{safe_symbol}_1m_{year}.csv"
        input_path = os.path.join(save_dir, input_filename)

        output_filename = f"{safe_symbol}_10m_{year}.csv"
        output_path = os.path.join(save_dir, output_filename)

        if not os.path.exists(input_path):
            print(f"[SKIP] Source file not found: {input_filename}")
            continue

        if os.path.exists(output_path):
            print(f"[SKIP] Target file already exists: {output_filename}")
            continue

        print(f"\n[PROCESSING] {input_filename} -> {output_filename}")

        try:
            df = pd.read_csv(input_path)
            df['datetime'] = pd.to_datetime(df['datetime'])
            df.set_index('datetime', inplace = True)

            ohlcv_dict = {
                'open': 'first',
                'high': 'max',
                'low': 'min',
                'close': 'last',
                'volume': 'sum'
            }

            df_10m = df.resample('10min').apply(ohlcv_dict)

            df_10m.dropna(inplace = True)

            df_10m.to_csv(output_path)
            print(f"[DONE] Saved {output_filename} (Rows: {len(df_10m)})")

        except Exception as e:
            print(f"[ERROR] Failed to process {year}: {e}")
```

### MF-DFA Function

```python
def mfdfa(data_array, scales, q_list, poly_order = 2):
    """
    Fuction for MFDFA by using Pytorch

    Args:
        batch_data: (Batch_Size, Window_Size) tensor
    """
    if isinstance(data_array, torch.Tensor):
        data_array = data_array.detach().cpu().numpy()
    elif isinstance(data_array, (pd.DataFrame, pd.Series)):
        data_array = data_array.values
    data_array = np.array(data_array)

    if data_array.ndim == 1:
        data_array = data_array.reshape(1, -1)

    batch_size, N = data_array.shape
    profile = np.cumsum(data_array - np.mean(data_array, axis = 1, keepdims = True), axis = 1)

    F_q_s = []
    valid_scales = []

    for s in scales:
        if N // s < 4: continue
        valid_scales.append(s)
        N_s = N // s

        seg_f = profile[:, :N_s*s].reshape(batch_size, N_s, s)
        seg_b = profile[:, -N_s*s:].reshape(batch_size, N_s, s)
        segments = np.concatenate([seg_f, seg_b], axis = 1)

        x_axis = np.linspace(-1, 1, s)

        X_design = np.vstack([x_axis**i for i in range(poly_order+1)]).T
        pinv_X = np.linalg.pinv(X_design)

        beta = np.matmul(segments, pinv_X.T)
        trend = np.matmul(beta, X_design.T)

        resid = segments - trend
        F2 = np.mean(resid**2, axis = 2)
        F2 = np.maximum(F2, 1e-20)

        F_vals = []
        for q in q_list:
            if abs(q) < 1e-5:
                f_q = np.exp(0.5 * np.mean(np.log(F2), axis = 1))
            else:
                f_q = np.mean(F2**(q/2), axis = 1)**(1/q)
            F_vals.append(f_q)

        F_q_s.append(np.stack(F_vals, axis = 1))

    if not F_q_s: return None, None

    F_output = np.stack(F_q_s, axis = 2)

    return np.array(valid_scales), F_output

def mfdfa_features(price_series, device, window, step):
    print(f"[INFO] Extracting MFDFA Features (Window: {window})")

    if isinstance(price_series, pd.DataFrame):
        price_series = price_series.iloc[:, 0]

    log_ret = np.log(price_series / price_series.shift(1)).dropna()
    values = log_ret.values

    windows = []
    indices = []

    BATCH_SIZE =   64
    feature_list = []

    q_list = np.linspace(-5, 5, 15)
    scales = np.unique(np.logspace(np.log10(10), np.log10(window//4), 10).astype(int))

    for t in tqdm(range(window, len(values), step)):
        seg = values[t-window : t]
        windows.append(seg)
        indices.append(log_ret.index[t])

        if len(windows) >= BATCH_SIZE:
            batch_tensor = torch.tensor(np.array(windows), dtype = torch.float32).to(device)
            s_vals, F_tensor = mfdfa(batch_tensor, scales, q_list)

            if F_tensor is not None:
                F_np = F_tensor
                log_s = np.log(s_vals)

                for i in range(len(F_np)):
                    h_q = []
                    for i_q in range(len(q_list)):
                        slope, _, _, _, _ = linregress(log_s, np.log(F_np[i, i_q, :]))
                        h_q.append(slope)

                    h_q = np.array(h_q)
                    delta_H = np.max(h_q) - np.min(h_q)
                    q_arr = np.array(q_list)
                    tau = q_arr * h_q - 1
                    alpha = np.gradient(tau, q_arr)

                    delta_alpha = np.max(alpha) - np.min(alpha)
                    alpha_0 = alpha[len(q_list)//2]

                    feature_list.append([delta_H, delta_alpha, alpha_0])

            windows = []
    if len(feature_list) != len(indices):
        min_len = min(len(feature_list), len(indices))
        feature_list = feature_list[:min_len]
        indices = indices[:min_len]

    return pd.DataFrame(feature_list, columns = ['delta_H', 'delta_alpha', 'alpha_0'], index = indices).dropna()


def compute_spectrum(price_series, q_list):
    log_ret = np.log(price_series / price_series.shift(1)).dropna()
    values = log_ret.values

    N = len(values)
    min_scale = 30
    max_scale = N // 10
    scales = np.unique(np.logspace(np.log10(min_scale), np.log10(max_scale), 30).astype(int))

    s_vals, F_batch = mfdfa(values, scales, q_list, poly_order = 2)

    if F_batch is None: return None

    F_np = F_batch[0]
    log_s = np.log(s_vals)
    h_q = []

    for i_q in range(len(q_list)):
        slope, _, _, _, _ = linregress(log_s, np.log(F_np[i_q, :]))
        h_q.append(slope)

    h_q = np.array(h_q)
    delta_H = np.max(h_q) - np.min(h_q)
    q_arr = np.array(q_list)
    tau = q_arr * h_q - 1
    alpha = np.gradient(tau, q_arr)
    delta_alpha = np.max(alpha) - np.min(alpha)
    f_alpha = q_arr * alpha - tau
    alpha_0 = alpha[len(q_list)//2]

    return{
        'scales': s_vals, 'F_q_s': F_np, 'H_q': h_q,
        'tau_q': tau, 'alpha': alpha, 'f_alpha': f_alpha,
        'delta_H': delta_H, 'delta_alpha': delta_alpha,
        'alpha_0': alpha_0
    }
```

### AI Model

```python
# 새로운 RV를 정의 (OHLCV 기반)
def ohlcv_rv(df, window = 144):
    T = window
    kappa = 1.34

    k = (kappa - 1) / (kappa + (T + 1) / (T - 1))

    o_t = np.log(df['open'] / df['close'].shift(1))

    c_t = np.log(df['close'] / df['open'])

    rsy = (np.log(df['high'] / df['close']) * np.log(df['high'] / df['open']) + \
           np.log(df['low'] / df['close']) * np.log(df['low'] / df['open']))
    
    sigma_o = o_t.rolling(window = T).var()
    sigma_c = c_t.rolling(window = T).var()

    sigma_rsy = rsy.rolling(window = T).mean()

    daily_variance = (sigma_o + k * sigma_c + (1 - k) * sigma_rsy) * T

    df['RV_OHLCV'] = np.sqrt(daily_variance)
    
    return df

class AsymmetricResidualLoss(nn.Module):
    def __init__(self, penalty_weight = 3.0):
        super(AsymmetricResidualLoss, self).__init__()
        self.penalty = penalty_weight
        
    def forward(self, y_pred, y_true):
        error = y_true - y_pred
        loss = torch.where(error > 0, (error * self.penalty)**2, error ** 2)
        return loss.mean()

class FlattenLSTM(nn.Module):
    def __init__(self, input_dim, hidden_dim = 64, num_layers = 2, dropout = 0.0):
        super(FlattenLSTM, self).__init__()
        eff_dropout = dropout if num_layers > 1 else 0.0
        self.lstm = nn.LSTM(input_dim, hidden_dim, num_layers, batch_first = True, dropout = eff_dropout)
        self.attention = nn.Sequential(nn.Linear(hidden_dim, 1), nn.Tanh())
        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x):
        out, _ = self.lstm(x)
        attn_weights = torch.softmax(self.attention(out), dim = 1)
        context = torch.sum(attn_weights * out, dim = 1)
        return self.fc(context)

class FlattenBiLSTM(nn.Module):
    def __init__(self, input_dim, hidden_dim = 64, num_layers = 2, dropout = 0.0):
        super(FlattenBiLSTM, self).__init__()        
        eff_dropout = dropout if num_layers > 1 else 0.0
        self.lstm = nn.LSTM(
            input_dim,
            hidden_dim,
            num_layers,
            batch_first = True,
            dropout = eff_dropout,
            bidirectional = True
        )        
        self.attention = nn.Sequential(nn.Linear(hidden_dim * 2, 1), nn.Tanh())
        self.fc = nn.Linear(hidden_dim * 2, 1)

    def forward(self, x):
        out, _ = self.lstm(x)
        attn_weights = torch.softmax(self.attention(out), dim=1)
        context = torch.sum(attn_weights * out, dim=1)
        return self.fc(context)

class FlattenGRU(nn.Module):
    def __init__(self, input_dim, hidden_dim = 64, num_layers = 2, dropout = 0.0):
        super(FlattenGRU, self).__init__()
        eff_dropout = dropout if num_layers > 1 else 0.0
        self.gru = nn.GRU(input_dim, hidden_dim, num_layers, batch_first = True, dropout = eff_dropout)
        self.attention = nn.Sequential(nn.Linear(hidden_dim, 1), nn.Tanh())
        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x):
        out, _ = self.gru(x)
        attn_weights = torch.softmax(self.attention(out), dim=1)
        context = torch.sum(attn_weights * out, dim=1)
        return self.fc(context)
    
class FlattenBiGRU(nn.Module):
    def __init__(self, input_dim, hidden_dim = 64, num_layers = 2, dropout = 0.0):
        super(FlattenBiGRU, self).__init__()
        eff_dropout = dropout if num_layers > 1 else 0.0
        self.gru = nn.GRU(
            input_dim, hidden_dim, num_layers,
            batch_first = True, bidirectional = True, dropout = eff_dropout
        )
        self.attention = nn.Sequential(nn.Linear(hidden_dim * 2, 1), nn.Tanh())
        self.fc = nn.Linear(hidden_dim * 2, 1)

    def forward(self, x):
        out, _ = self.gru(x)
        attn_weights = torch.softmax(self.attention(out), dim=1)
        context = torch.sum(attn_weights * out, dim=1)
        return self.fc(context)

def prepare_flatten_features(df_base, mfdfa_df, lags=144):
    print(f"Data Preprocessing: Creating Lags={lags} (Safe & Optimized)")

    # ---------------------------------------------------------
    # 1. MF-DFA 전처리
    # ---------------------------------------------------------
    if df_base.index.tz is not None: df_base.index = df_base.index.tz_localize(None)
    
    mfdfa_cols = ['delta_H', 'delta_alpha', 'alpha_0']
    mfdfa_subset = mfdfa_df[mfdfa_cols].copy()

    if mfdfa_subset.index.tz is not None: mfdfa_subset.index = mfdfa_subset.index.tz_localize(None)


    mfdfa_daily = mfdfa_subset.resample('1D').last()

    mfdfa_daily.index = pd.to_datetime(mfdfa_daily.index).normalize()

    mfdfa_shifted = mfdfa_daily.shift(1)

    # ---------------------------------------------------------
    # 2. 데이터 병합
    # ---------------------------------------------------------
    df_base['date_key'] = df_base.index.normalize()

    merged_data = pd.merge(df_base, mfdfa_shifted,
                           left_on='date_key', right_index=True,
                           how='left')

    merged_data.drop(columns=['date_key'], inplace=True)

    merged_data[mfdfa_cols] = merged_data[mfdfa_cols].fillna(method='ffill')

    # ---------------------------------------------------------
    # 3. 변수 생성 및 메모리 최적화
    # ---------------------------------------------------------
    merged_data['Future_Vol_24h'] = merged_data['RV_OHLCV'].shift(-144)

    merged_data['Base_Vol'] = merged_data['gm_vol'].fillna(method='ffill')
    
    # merged_data['Target_Residual'] = merged_data['Future_Vol_24h'] - (merged_data['Base_Vol'])
    merged_data['Target_Residual'] = np.log((merged_data['Future_Vol_24h'] / merged_data['Base_Vol']))
    
    merged_data['Past_GARCH_RV'] = merged_data['Base_Vol']
    merged_data['Past_Residual'] = np.log((merged_data['RV_OHLCV'] / merged_data['Base_Vol']))

    cols_needed = ['Target_Residual', 'Base_Vol', 'Future_Vol_24h', 
                   'RV_OHLCV', 'return', 'Past_Residual'] + mfdfa_cols
    data = merged_data[cols_needed].copy()

    # float32 변환 (메모리 절약)
    for col in data.select_dtypes(include=['float64']).columns:
        data[col] = data[col].astype('float32')

    data.dropna(inplace=True)

    # ---------------------------------------------------------
    # 4. Lag Feature 생성
    # ---------------------------------------------------------
    base_features = ['RV_OHLCV', 'return', 'Past_Residual'] + mfdfa_cols
    feature_dfs = [data]
    all_feature_cols = []

    for lag in tqdm(range(1, lags + 1), desc="Generating Lags"):
        shifted = data[base_features].shift(lag)
        shifted.columns = [f"{col}_lag{lag}" for col in base_features]
        feature_dfs.append(shifted)
        all_feature_cols.extend(shifted.columns)

    del merged_data, mfdfa_shifted, mfdfa_daily
    gc.collect()

    data_final = pd.concat(feature_dfs, axis=1)
    data_final.dropna(inplace=True)

    print(f"Final Data shape: {data_final.shape}")

    return data_final, all_feature_cols

def run_rolling_simulation(full_data, feature_cols, model_type = 'LSTM', train_window = 4320):
    print(f"\n[INFO] {model_type} Rolling Simulation Start")

    X_full = full_data[feature_cols].values

    full_data['GARCH_Daily'] = full_data['Base_Vol'] * 12

    full_data['Residual'] = full_data['Future_Vol_24h'] - full_data['GARCH_Daily']
    y_full = full_data[['Residual']].values

    start_idx = train_window
    total_len = len(full_data)

    scaler_X = MinMaxScaler((-1, 1))
    scaler_X.fit(X_full[:start_idx])
    X_scaled = scaler_X.transform(X_full)

    scaler_y = MinMaxScaler((-1, 1))
    scaler_y.fit(y_full[:start_idx])
    y_scaled = scaler_y.transform(y_full)

    model_type = model_type.upper()

    is_deep_learning = model_type in ['LSTM', 'BI-LSTM', 'GRU', 'BI-GRU']
    if is_deep_learning:
        X_tensor = torch.FloatTensor(X_scaled).to(DEVICE)
        y_tensor = torch.FloatTensor(y_scaled).to(DEVICE)

        X_tensor = X_tensor.unsqueeze(1)

        input_dim =len(feature_cols)
        hidden_dim = 64
        num_layers = 2

        if model_type == 'LSTM':
            model = FlattenLSTM(input_dim, hidden_dim, num_layers).to(DEVICE)
        elif model_type == 'BI-LSTM':
            model = FlattenBiLSTM(input_dim, hidden_dim, num_layers).to(DEVICE)
        elif model_type == 'GRU':
            model = FlattenGRU(input_dim, hidden_dim, num_layers).to(DEVICE)
        elif model_type == 'BI-GRU':
            model = FlattenBiGRU(input_dim, hidden_dim, num_layers).to(DEVICE)
        else:
            raise ValueError(f"Not supported model type: {model_type}")

        optimizer = torch.optim.Adam(model.parameters(), lr = 0.001)
        loss_fn = nn.HuberLoss(delta = 1.0).to(DEVICE)
    else:
        pass

    print(f" -> 총 데이터: {total_len}, 시뮬레이션 시작 인덱스: {start_idx}")

    predictions = []
    indices = []
    loss_history = []

    end_idx = len(full_data)

    sim_start = start_idx

    pbar = tqdm(range(sim_start, end_idx), desc = f"{model_type} Rolling Update")

    for t in pbar:
        if is_deep_learning:
            model.train()

            X_batch = X_tensor[t - train_window : t]
            y_batch = y_tensor[t - train_window : t]

            optimizer.zero_grad()
            pred_train = model(X_batch)
            loss = loss_fn(pred_train, y_batch)
            loss.backward()
            optimizer.step()

            loss_history.append(loss.item())

            model.eval()
            with torch.no_grad():
                X_input = X_tensor[t].unsqueeze(0)
                pred_val = model(X_input).item()
        else:
            ridge_model = Ridge(alpha = 1.0)
            X_train = X_scaled[t - train_window : t]
            y_train = y_scaled[t - train_window : t]
            ridge_model.fit(X_train, y_train)

            X_curr = X_scaled[t].reshape(1, -1)
            pred_val = ridge_model.predict(X_curr).item()

        predictions.append(pred_val)
        indices.append(full_data.index[t])

    pred_array = np.array(predictions).reshape(-1, 1)
    pred_rescaled = scaler_y.inverse_transform(pred_array).flatten()

    result_df = pd.DataFrame(index = indices)
    result_df['Predicted_Residual'] = pred_rescaled

    final_df = pd.merge(result_df, full_data[['Residual', 'Base_Vol', 'Future_Vol_24h', 'GARCH_Daily']],
                        left_index = True, right_index = True, how = 'inner')

    final_df['Hybrid_Vol'] = final_df['GARCH_Daily'] + final_df['Predicted_Residual']

    return final_df, loss_history

class FlattenLSTM_tuned(nn.Module):
    def __init__(self, input_dim, hidden_dim, num_layers, dropout = 0.0):
        super(FlattenLSTM_tuned, self).__init__()
        eff_dropout = dropout if num_layers > 1 else 0.0
        self.lstm = nn.LSTM(input_dim, hidden_dim, num_layers, batch_first = True, dropout = eff_dropout)
        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x):
        _, (hn, _) = self.lstm(x)
        out = self.fc(hn[-1])
        return out.squeeze()

class FlattenBiLSTM_tuned(nn.Module):
    def __init__(self, input_dim, hidden_dim, num_layers, dropout = 0.0):
        super(FlattenBiLSTM_tuned, self).__init__()
        eff_dropout = dropout if num_layers > 1 else 0.0
        self.bilstm = nn.LSTM(input_dim, hidden_dim, num_layers, batch_first = True, dropout = eff_dropout, bidirectional = True)
        self.fc = nn.Linear(hidden_dim * 2, 1)

    def forward(self, x):
        _, (hn, _) = self.bilstm(x)
        hn_cat = torch.cat((hn[-2], hn[-1]), dim = 1)
        out = self.fc(hn_cat)
        return out.squeeze()

class FlattenGRU_tuned(nn.Module):
    def __init__(self, input_dim, hidden_dim, num_layers, dropout = 0.0):
        super(FlattenGRU_tuned, self).__init__()
        eff_dropout = dropout if num_layers > 1 else 0.0
        self.gru = nn.GRU(input_dim, hidden_dim, num_layers, batch_first = True, dropout = eff_dropout)
        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x):
        _, hn = self.gru(x)
        out = self.fc(hn[-1])
        return out.squeeze()

class FlattenBiGRU_tuned(nn.Module):
    def __init__(self, input_dim, hidden_dim, num_layers, dropout = 0.0):
        super(FlattenBiGRU_tuned, self).__init__()
        eff_dropout = dropout if num_layers > 1 else 0.0
        self.bigru = nn.GRU(input_dim, hidden_dim, num_layers, batch_first = True, dropout = eff_dropout, bidirectional = True)
        self.fc = nn.Linear(hidden_dim * 2, 1)

    def forward(self, x):
        _, hn = self.bigru(x)
        hn_cat = torch.cat((hn[-2], hn[-1]), dim = 1)
        out = self.fc(hn_cat)
        return out.squeeze()

def run_rolling_simulation_tuned(full_data, feature_cols, model_type = 'LSTM', train_window = 4320, params = None, custom_range = None):
    """
    Run Rolling Simulation with Tuned Models
    """
    model_type = model_type.upper()

    if params is None: params = {}

    hidden_dim = params.get('hidden_dim', 64)
    num_layers = params.get('num_layers', 1)
    lr = params.get('learning_rate', 0.001)
    dropout = params.get('dropout', 0.0)
    alpha_val = params.get('alpha', 1.0)

    print(f"\n[Simulation Tuned] {model_type} | Config: {params}")

    X_full = full_data[feature_cols].values

    if 'GARCH_Daily' not in full_data.columns:
        full_data['GARCH_Daily'] = full_data['Base_Vol'] * 12
        full_data['Residual'] = full_data['Future_Vol_24h'] - full_data['GARCH_Daily']

    y_full = full_data[['Residual']].values

    scaler_X = MinMaxScaler((-1, 1))
    scaler_X.fit(X_full[:train_window])
    X_scaled = scaler_X.transform(X_full)

    scaler_y = MinMaxScaler((-1, 1))
    scaler_y.fit(y_full[:train_window])
    y_scaled = scaler_y.transform(y_full)

    if custom_range:
        start_idx, end_idx = custom_range
        if start_idx < train_window:
            start_idx = train_window
    else:
        start_idx = train_window
        end_idx = len(full_data)

    predictions = []
    indices = []
    loss_history = []

    input_dim = len(feature_cols)

    is_deep_learning = model_type in ['LSTM', 'BI-LSTM', 'GRU', 'BI-GRU']

    if is_deep_learning:
        X_tensor = torch.FloatTensor(X_scaled).to(DEVICE).unsqueeze(1)
        y_tensor = torch.FloatTensor(y_scaled).to(DEVICE)

        if model_type == 'LSTM':
            model = FlattenLSTM_tuned(input_dim, hidden_dim, num_layers, dropout).to(DEVICE)
        elif model_type == 'BI-LSTM':
            model = FlattenBiLSTM_tuned(input_dim, hidden_dim, num_layers, dropout).to(DEVICE)
        elif model_type == 'GRU':
            model = FlattenGRU_tuned(input_dim, hidden_dim, num_layers, dropout).to(DEVICE)
        elif model_type == 'BI-GRU':
            model = FlattenBiGRU_tuned(input_dim, hidden_dim, num_layers, dropout).to(DEVICE)

        optimizer = torch.optim.Adam(model.parameters(), lr = lr)
        loss_fn = nn.HuberLoss(delta = 1.0).to(DEVICE)

    pbar = tqdm(range(start_idx, end_idx), desc=f"{model_type} Rolling")
    for t in pbar:
        if is_deep_learning:
            model.train()
            X_batch = X_tensor[t - train_window : t]
            y_batch = y_tensor[t - train_window : t]

            optimizer.zero_grad()
            pred_train = model(X_batch)
            loss = loss_fn(pred_train, y_batch)
            loss.backward()
            optimizer.step()
            loss_history.append(loss.item())

            model.eval()
            with torch.no_grad():
                X_input = X_tensor[t].unsqueeze(0)
                pred_val = model(X_input).item()
        else:
            ridge_model = Ridge(alpha = alpha_val)
            X_train = X_scaled[t - train_window : t]
            y_train = y_scaled[t - train_window : t]
            ridge_model.fit(X_train, y_train)

            X_curr = X_scaled[t].reshape(1, -1)
            pred_val = ridge_model.predict(X_curr).item()

        predictions.append(pred_val)
        indices.append(full_data.index[t])

    pred_array = np.array(predictions).reshape(-1, 1)
    pred_rescaled = scaler_y.inverse_transform(pred_array).flatten()

    result_df = pd.DataFrame(index = indices)
    result_df['Predicted_Residual'] = pred_rescaled

    final_df = pd.merge(result_df, full_data[['Residual', 'Base_Vol', 'Future_Vol_24h', 'GARCH_Daily']],
                        left_index = True, right_index = True, how = 'inner')

    final_df['Hybrid_Vol'] = final_df['GARCH_Daily'] + final_df['Predicted_Residual']

    print(f"   [Debug] Predictions Std: {np.std(predictions):.6f}, Mean: {np.mean(predictions):.6f}")

    return final_df, loss_history

def run_direct_rv_simulation(full_data, feature_cols, model_type = 'LSTM', train_window = 4320):

    hidden_dim = 64
    num_layers = 2
    lr = 0.001
    dropout = 0.0
    alpha_val = 1.0

    model_type = model_type.upper()

    X_full = full_data[feature_cols].values
    y_full = full_data[['Future_Vol_24h']].values

    scaler_X = MinMaxScaler((-1, 1))
    scaler_X.fit(X_full[:train_window])
    X_scaled = scaler_X.transform(X_full)

    scaler_y = MinMaxScaler((-1, 1))
    scaler_y.fit(y_full[:train_window])
    y_scaled = scaler_y.transform(y_full)

    start_idx = train_window
    end_idx = len(full_data)

    is_dl = model_type in ['LSTM', 'BI-LSTM', 'GRU', 'BI-GRU']

    if is_dl:
        X_tensor = torch.FloatTensor(X_scaled).to(DEVICE).unsqueeze(1)
        y_tensor = torch.FloatTensor(y_scaled).to(DEVICE)

        input_dim = len(feature_cols)

        if model_type == 'LSTM':
            model = FlattenLSTM_tuned(input_dim, hidden_dim, num_layers, dropout).to(DEVICE)
        elif model_type == 'BI-LSTM':
            model = FlattenBiLSTM_tuned(input_dim, hidden_dim,num_layers, dropout).to(DEVICE)
        elif model_type == 'GRU':
            model = FlattenGRU_tuned(input_dim, hidden_dim,num_layers, dropout).to(DEVICE)
        elif model_type == 'BI-GRU':
            model = FlattenBiGRU_tuned(input_dim, hidden_dim,num_layers, dropout).to(DEVICE)

        optimizer = torch.optim.Adam(model.parameters(), lr = lr)
        loss_fn = nn.HuberLoss(delta = 1.0).to(DEVICE)

    predictions = []
    indices = []

    pbar = tqdm(range(start_idx, end_idx), desc=f"{model_type} Rolling")
    for t in pbar:
        if is_dl:
            model.train()
            X_batch = X_tensor[t - train_window : t]
            y_batch = y_tensor[t - train_window : t]

            optimizer.zero_grad()
            pred_train = model(X_batch)
            loss = loss_fn(pred_train, y_batch)
            loss.backward()
            optimizer.step()

            model.eval()
            with torch.no_grad():
                X_input = X_tensor[t].unsqueeze(0)
                pred_val = model(X_input).item()
        else:
            ridge_model = Ridge(alpha = alpha_val)
            X_train = X_scaled[t - train_window : t]
            y_train = y_scaled[t - train_window : t]
            ridge_model.fit(X_train, y_train)

            X_curr = X_scaled[t].reshape(1, -1)
            pred_val = ridge_model.predict(X_curr).item()

        predictions.append(pred_val)
        indices.append(full_data.index[t])

    pred_array = np.array(predictions).reshape(-1, 1)
    pred_rescaled = scaler_y.inverse_transform(pred_array).flatten()

    result_df = pd.DataFrame(index = indices)
    result_df['Predicted_RV'] = pred_rescaled

    final_df = pd.merge(result_df, full_data[['Future_Vol_24h']],
                        left_index = True, right_index = True, how = 'inner')
    if is_dl:
        del model, optimizer, X_tensor, y_tensor
        torch.cuda.empty_cache()

    return final_df
```

### GARCH-MIDAS Model

```python
def get_beta_weights(K, w1, w2):
    k_vals = np.linspace(1e-6, 1.0 - 1e-6, K)
    weights = k_vals**(w1 - 1) * (1 - k_vals)**(w2 - 1)
    return weights / weights.sum()

def garch_midas(train_df, test_df, daily_rv, df_x, midas_k = 30):
    print("[INFO] Running GARCH-MIDAS")

    full_df = pd.concat([train_df, test_df])

    train_end_idx = train_df.index[-1]

    daily_rv_scaled = daily_rv.copy()
    if daily_rv_scaled['rv'].mean() < 0.1:
        daily_rv_scaled['rv'] = daily_rv_scaled['rv'] * 10000

    combined_daily = pd.merge(daily_rv_scaled[['rv']], df_x, left_index = True, right_index = True, how = 'inner')

    weights = get_beta_weights(midas_k, 1.0, 5.0)

    combined_daily['midas_rv'] = combined_daily['rv'].rolling(midas_k).apply(
        lambda x: np.dot(x, weights[::-1]), raw = True
    ).shift(1)

    x_col_name = df_x.columns[0]
    combined_daily['exog_x'] = combined_daily[x_col_name].rolling(midas_k).apply(
        lambda x: np.dot(x, weights[::-1]), raw = True
    ).shift(1)

    combined_daily.dropna(inplace = True)

    merged = full_df.join(combined_daily[['midas_rv', 'exog_x']], how = 'left')
    merged[['midas_rv', 'exog_x']] = merged[['midas_rv', 'exog_x']].fillna(method='ffill')
    merged.dropna(inplace=True)

    if merged['return'].abs().mean() < 0.1:
        merged['return_scaled'] = merged['return'] * 100
    else:
        merged['return_scaled'] = merged['return']

    mask_train = merged.index <= train_end_idx

    train_ret = merged.loc[mask_train, 'return_scaled'].values
    train_midas = merged.loc[mask_train, 'midas_rv'].values
    train_exog = merged.loc[mask_train, 'exog_x'].values
    T_train = len(train_ret)


    def objective(params):
        alpha, beta, m, theta, beta_x = params

        if alpha + beta >= 0.999 or alpha < 0 or beta < 0:
            return 1e10

        tau = (m**2) + (theta**2 * train_midas) + (beta_x**2 * train_exog)
        tau = np.maximum(tau, 1e-12)

        g = np.ones(T_train)
        e2_over_tau = (train_ret**2) / tau
        omega = (1 - alpha - beta)

        g[0] = 1.0
        for t in range(1, T_train):
            g[t] = omega + alpha * e2_over_tau[t-1] + beta * g[t-1]

        h = tau * g
        llh = 0.5 * (np.log(2 * np.pi) + np.log(h) + (train_ret**2) / h)
        return np.sum(llh)

    initial_guess = [0.05, 0.90, 0.1, 0.3, 0.1]
    bounds = [(0, 1), (0, 1), (None, None), (None, None), (None, None)]

    try:
        res = minimize(objective, initial_guess, method = 'L-BFGS-B', bounds = bounds)
        best_params = res.x
        success = res.success
        msg = res.message
    except:
        best_params = initial_guess
        success = False
        msg = "Optimization Crash"

    alpha, beta, m ,theta, beta_x = best_params

    full_ret = merged['return_scaled'].values
    full_midas = merged['midas_rv'].values
    full_exog = merged['exog_x'].values
    T_full = len(full_ret)

    tau_full = (m**2) + (theta**2 * full_midas) + (beta_x**2 * full_exog)
    tau_full = np.maximum(tau_full, 1e-12)

    g_full = np.ones(T_full)
    e2_over_tau_full = (full_ret**2) / tau_full
    omega = (1 - alpha - beta)

    for t in range(1, T_full):
        g_full[t] = omega + alpha * e2_over_tau_full[t-1] + beta * g_full[t-1]

    merged['gm_vol'] = np.sqrt(tau_full * g_full)

    mask_test = merged.index > train_end_idx
    test_result = merged.loc[mask_test].copy()

    print("-" * 40)
    print(f"[DEBUG] Optimization Result: {msg} (Success: {success})")
    print(f"[DEBUG] Best Parameters:")
    print(f"   - Alpha (GARCH) : {alpha:.6f}")
    print(f"   - Beta  (GARCH) : {beta:.6f}")
    print(f"   - m     (Const) : {m:.6f}  -> (m^2 used: {m**2:.8f})")
    print(f"   - Theta (RV_Wgt): {theta:.6f}  -> (theta^2 used: {theta**2:.8f})")
    print(f"   - W_Exog({x_col_name}): {beta_x:.6f}  -> (w^2 used: {beta_x**2:.8f})")

    print(f"[DEBUG] Predicted RV (gm_vol) Stats (Test Period):")
    if len(test_result) > 0:
        print(f"   - Mean: {test_result['gm_vol'].mean():.6f}")
        print(f"   - Min : {test_result['gm_vol'].min():.6f}")
        print(f"   - Max : {test_result['gm_vol'].max():.6f}")
        print(f"   - First 5 values: {test_result['gm_vol'].values[:5]}")
    else:
        print("   - [Warning] Test result is empty.")
    print("-" * 60)

    return test_result

def monthly_garch_midas(df_target, mfdfa_df, exog_col = None, midas_k = 30):
    """
    Predict Realized Volatility by using Base Model (GARCH_MIDAS)
    Don't use future data, use past data to estimate parameters
    """
    print(f"[INFO] Base Model Running (Exog: {exog_col if exog_col else 'None'})")

    monthly_results = []

    month_starts = df_target.resample('MS').first().index

    min_train_month = 2

    daily_rv_all = (df_target['return']**2).resample('1D').sum().to_frame('rv')

    mfdfa_df.index = pd.to_datetime(mfdfa_df.index).normalize()

    if exog_col is None:
        df_x_full = pd.DataFrame(0, index = mfdfa_df.index, columns = ['exog_x'])
    else:
        if exog_col not in mfdfa_df.columns:
            raise KeyError(f"Not Exist '{exog_col}' column in MF-DFA data")
        df_x_full = mfdfa_df[[exog_col]].copy()
        df_x_full.columns = ['exog_x']

    df_x_full = df_x_full.resample('1D').mean()

    print(f"   -> RV 데이터 기간: {daily_rv_all.index.min()} ~ {daily_rv_all.index.max()}")
    print(f"   -> MF-DFA 데이터 기간: {df_x_full.index.min()} ~ {df_x_full.index.max()}")

    # 교집합 확인
    common_idx = daily_rv_all.index.intersection(df_x_full.index)
    print(f"   -> 겹치는 날짜 수: {len(common_idx)}일")

    if len(common_idx) < 100:
        raise ValueError(f"데이터 날짜가 서로 안 맞습니다! 겹치는 날짜가 {len(common_idx)}개 뿐입니다.")

    df_x_full = df_x_full.reindex(daily_rv_all.index).fillna(method = 'ffill').fillna(0)

    pbar = tqdm(range(min_train_month, len(month_starts)), desc = "Monthly Re-fitting")

    for i in pbar:
        current_month = month_starts[i]
        next_month = month_starts[i + 1] if i+1 < len(month_starts) else None

        train_slice = df_target[:current_month]

        if next_month:
            test_slice = df_target[current_month:next_month]
        else:
            test_slice = df_target[current_month:]

        try:
            pred_df = garch_midas(
                train_slice, test_slice, daily_rv_all, df_x_full, midas_k
            )
            monthly_results.append(pred_df)
        except Exception as e:
            print(f"\n Error at {current_month.strftime('%Y-%m')}: {e}")
            pass

    if len(monthly_results) > 0:
        final_gm_df = pd.concat(monthly_results).sort_index()
        print(f"Complete {len(final_gm_df)} Created")
        return final_gm_df
    else:
        raise ValueError("No result by Failing GARCH0-MDIAS Model Training")

def garch_midas_tuned(train_df, test_df, daily_rv, df_x, midas_k = 30):

    full_df = pd.concat([train_df, test_df])

    train_end_idx = train_df.index[-1]
    
    daily_rv_scaled = daily_rv.copy()
    daily_rv_scaled['rv'] = daily_rv_scaled['rv'] * 10000

    combined_daily = pd.merge(daily_rv_scaled[['rv']], df_x, left_index = True, right_index = True, how = 'inner')

    weights = get_beta_weights(midas_k, 1.0, 5.0)

    combined_daily['midas_rv'] = combined_daily['rv'].rolling(midas_k).apply(
        lambda x: np.dot(x, weights[::-1]), raw = True
    ).shift(1)

    x_col_name = df_x.columns[0]
    combined_daily['exog_x'] = combined_daily[x_col_name].rolling(midas_k).apply(
        lambda x: np.dot(x, weights[::-1]), raw = True
    ).shift(1)

    combined_daily.dropna(inplace = True)

    merged = full_df.copy()
    merged['date_key'] = merged.index.normalize()
    combined_daily.index = combined_daily.index.normalize()
    
    merged = merged.merge(combined_daily[['midas_rv', 'exog_x']], left_on='date_key', right_index=True, how='left')
    merged.drop(columns=['date_key'], inplace=True)
    
    merged[['midas_rv', 'exog_x']] = merged[['midas_rv', 'exog_x']].fillna(method='ffill')
    merged.dropna(inplace=True)

    merged['return_scaled'] = merged['return']

    mask_train = merged.index <= train_end_idx

    train_ret = merged.loc[mask_train, 'return_scaled'].values
    train_midas = merged.loc[mask_train, 'midas_rv'].values
    train_exog = merged.loc[mask_train, 'exog_x'].values
    T_train = len(train_ret)

    def objective(params):
        alpha, beta, m, theta, beta_x = params

        if alpha + beta >= 0.999 or alpha < 0 or beta < 0:
            return 1e10

        tau = (m**2) + (theta**2 * train_midas) + (beta_x**2 * train_exog)
        tau = np.maximum(tau, 1e-12)
        
        tau_10m = tau / 144.0

        g = np.ones(T_train)
        e2_over_tau = (train_ret**2) / tau_10m
        omega = (1 - alpha - beta)

        g[0] = 1.0
        for t in range(1, T_train):
            g[t] = omega + alpha * e2_over_tau[t-1] + beta * g[t-1]

        h = tau_10m * g
        llh = 0.5 * (np.log(2 * np.pi) + np.log(h) + (train_ret**2) / h)
        return np.sum(llh)

    initial_guess = [0.05, 0.90, 0.1, 0.3, 0.1]
    bounds = [(0, 1), (0, 1), (None, None), (None, None), (None, None)]

    try:
        res = minimize(objective, initial_guess, method = 'L-BFGS-B', bounds = bounds)
        best_params = res.x
        msg = res.message
        success = res.success
    except Exception as e:
        best_params = initial_guess
        msg = f"Optimization Failed: {str(e)}"
        success = False

    alpha, beta, m ,theta, beta_x = best_params

    full_ret = merged['return_scaled'].values
    full_midas = merged['midas_rv'].values
    full_exog = merged['exog_x'].values
    T_full = len(full_ret)

    tau_full = (m**2) + (theta**2 * full_midas) + (beta_x**2 * full_exog)
    tau_full = np.maximum(tau_full, 1e-12)
    tau_full_10m = tau_full / 144.0

    g_full = np.ones(T_full)
    e2_over_tau_full = (full_ret**2) / tau_full_10m
    omega = (1 - alpha - beta)

    for t in range(1, T_full):
        g_full[t] = omega + alpha * e2_over_tau_full[t-1] + beta * g_full[t-1]

    merged['gm_vol'] = (np.sqrt(tau_full_10m * g_full) / 100) * 12

    mask_test = merged.index >= (train_end_idx - pd.Timedelta(days = 1))
    test_result = merged.loc[mask_test].copy()
    
    print("-" * 40)
    print(f"[DEBUG] Optimization Result: {msg} (Success: {success})")
    print(f"[DEBUG] Best Parameters:")
    print(f"   - Alpha (GARCH) : {alpha:.6f}")
    print(f"   - Beta  (GARCH) : {beta:.6f}")
    print(f"   - m     (Const) : {m:.6f}  -> (m^2 used: {m**2:.8f})")
    print(f"   - Theta (RV_Wgt): {theta:.6f}  -> (theta^2 used: {theta**2:.8f})")
    print(f"   - W_Exog({x_col_name}): {beta_x:.6f}  -> (w^2 used: {beta_x**2:.8f})")

    print(f"[DEBUG] Predicted RV (gm_vol) Stats (Test Period):")
    if len(test_result) > 0:
        print(f"   - Mean: {test_result['gm_vol'].mean():.6f}")
        print(f"   - Min : {test_result['gm_vol'].min():.6f}")
        print(f"   - Max : {test_result['gm_vol'].max():.6f}")
        print(f"   - First 5 values: {test_result['gm_vol'].values[:5]}")
    else:
        print("   - [Warning] Test result is empty.")
    print("-" * 60)

    return merged

def monthly_garch_midas_rolling_tuned(df_target, mfdfa_df, exog_col = None, midas_k = 30):
    """
    Apply midas_k value to monthly_garch_midas function to find optimal k
    """

    df_target = df_target.copy()
    if df_target.index.tz is not None:
        df_target.index = df_target.index.tz_localize(None)

    mfdfa_df = mfdfa_df.copy()
    mfdfa_df.index = pd.to_datetime(mfdfa_df.index).normalize()
    if mfdfa_df.index.tz is not None:
        mfdfa_df.index = mfdfa_df.index.tz_localize(None)

    WINDOW_MONTH = 2
    WINDOW_SIZE = int(WINDOW_MONTH * 30 * 144)

    print(f"[INFO] Base Model Running (Exog: {exog_col if exog_col else 'None'})")
    print(f"   -> [GARCH-MIDAS] Running with midas_k = {midas_k}")

    monthly_results = []

    month_starts = df_target.resample('MS').first().index

    min_start_idx = WINDOW_MONTH
    if min_start_idx >= len(month_starts):
        return pd.DataFrame()

    daily_rv_all = (df_target['return']**2).resample('1D').sum().to_frame('rv')

    mfdfa_df.index = pd.to_datetime(mfdfa_df.index).normalize()

    if exog_col is None:
        df_x_full = pd.DataFrame(0, index = daily_rv_all.index, columns = ['exog_x'])
    else:
        if exog_col not in mfdfa_df.columns:
            raise KeyError(f"Not Exist '{exog_col}' column in MF-DFA data")
        df_x_full = mfdfa_df[[exog_col]].copy()
        df_x_full.columns = ['exog_x']

    df_x_full = df_x_full.resample('1D').mean()

    # 교집합 확인
    common_idx = daily_rv_all.index.intersection(df_x_full.index)
    print(f"   -> 겹치는 날짜 수: {len(common_idx)}일")

    if len(common_idx) < 100:
        raise ValueError(f"데이터 날짜가 서로 안 맞습니다! 겹치는 날짜가 {len(common_idx)}개 뿐입니다.")

    df_x_full = df_x_full.reindex(daily_rv_all.index).fillna(method = 'ffill').fillna(0)

    pbar = tqdm(range(min_start_idx, len(month_starts)), desc = "Monthly Re-fitting")

    for i in pbar:
        current_month = month_starts[i]
        next_month = month_starts[i + 1] if i+1 < len(month_starts) else None

        try:
            curr_loc = df_target.index.searchsorted(current_month)
            if isinstance(curr_loc, slice): curr_loc = curr_loc.start
        except KeyError: continue
        start_loc = max(0, curr_loc - WINDOW_SIZE)

        train_slice = df_target.iloc[start_loc : curr_loc]

        if next_month:
            next_loc = df_target.index.searchsorted(next_month)
            test_slice = df_target[curr_loc : next_loc]
        else:
            test_slice = df_target[curr_loc:]

        try:
            pred_df = garch_midas_tuned(
                train_slice, test_slice, daily_rv_all, df_x_full, midas_k
            )
            monthly_results.append(pred_df)
        except Exception as e:
            print(f"\n Error at {current_month.strftime('%Y-%m')}: {e}")
            pass

    if len(monthly_results) > 0:
        final_gm_df = pd.concat(monthly_results).sort_index()
        print(f"Complete {len(final_gm_df)} Created")
        return final_gm_df
    else:
        raise ValueError("No result by Failing GARCH0-MDIAS Model Training")
```

```python
# =========================================================
# 1. 롤링 윈도우 전용 GARCH-MIDAS 함수
# =========================================================
def monthly_garch_midas_rolling(df_target, mfdfa_df, exog_col=None, midas_k=30, window_months=3):
    """
    window_months: 학습에 사용할 과거 데이터의 개월 수 (Rolling Window Size)
    """

    # 1개월 = 30일 * 144개(10분봉) = 4320개
    WINDOW_SIZE = int(window_months * 30 * 144)

    # 결과 저장용 리스트
    monthly_results = []

    # 월별 리밸런싱 지점 (매월 1일)
    month_starts = df_target.resample('MS').first().index

    # RV 및 Exog 데이터 준비 (기존 로직 동일)
    daily_rv_all = (df_target['return']**2).resample('1D').sum().to_frame('rv')
    mfdfa_df.index = pd.to_datetime(mfdfa_df.index).normalize()

    if exog_col is None:
        df_x_full = pd.DataFrame(0, index=mfdfa_df.index, columns=['exog_x'])
    else:
        if exog_col not in mfdfa_df.columns:
            raise KeyError(f"Not Exist '{exog_col}' column")
        df_x_full = mfdfa_df[[exog_col]].copy()
        df_x_full.columns = ['exog_x']

    df_x_full = df_x_full.resample('1D').mean()
    df_x_full = df_x_full.reindex(daily_rv_all.index).fillna(method='ffill').fillna(0)

    # 튜닝 시작 (최소 window_months 이후부터 시작)
    start_month_idx = window_months
    if start_month_idx >= len(month_starts):
        print(f"[Warning] 데이터가 너무 짧아서 {window_months}개월 윈도우를 적용할 수 없습니다.")
        return pd.DataFrame()

    # 진행바 설정
    pbar = tqdm(range(start_month_idx, len(month_starts)), desc=f"Rolling Window ({window_months}M)")

    for i in pbar:
        current_month = month_starts[i]
        next_month = month_starts[i + 1] if i + 1 < len(month_starts) else None

        # === [핵심] Rolling Window Slicing ===
        # 현재 달의 시작 위치(인덱스)를 찾음
        try:
            curr_loc = df_target.index.get_loc(current_month)
            # 단일 값이 아니라 슬라이스 객체일 경우 처리 (Duplicate index 방지)
            if isinstance(curr_loc, slice):
                curr_loc = curr_loc.start
        except KeyError:
            continue

        # 시작점 = 현재 - 윈도우 크기
        start_loc = max(0, curr_loc - WINDOW_SIZE)

        # 학습 데이터: [과거 윈도우 ~ 현재 직전]
        train_slice = df_target.iloc[start_loc : curr_loc]

        # 테스트 데이터: [현재 ~ 다음 달]
        if next_month:
            test_slice = df_target[current_month : next_month]
        else:
            test_slice = df_target[current_month:]

        # 데이터가 너무 적으면 스킵 (최소 20일치 이상)
        if len(train_slice) < 2880:
            continue

        try:
            # 튜닝된 GARCH-MIDAS 실행
            pred_df = garch_midas(
                train_slice, test_slice, daily_rv_all, df_x_full, midas_k=midas_k
            )
            monthly_results.append(pred_df)
        except Exception as e:
            # print(f" Error: {e}")
            pass

    if len(monthly_results) > 0:
        return pd.concat(monthly_results).sort_index()
    else:
        return pd.DataFrame()

# =========================================================
# 2. 윈도우 크기별 튜닝 실험 실행기
# =========================================================
def tune_rolling_window_size(df_10m, mfdfa_data):

    # 테스트할 윈도우 크기 (1개월, 2개월, 3개월)
    windows = [1, 2, 3]

    # 테스트할 시나리오 (가장 성능이 좋았던 Delta_H와 Benchmark만 비교)
    scenarios = {
        "Benchmark (No Exog)": None,
        "Delta_H (With MF)": "delta_H"
    }

    results = []
    lags = 144
    scale_factor = 12

    print("\n" + "="*60)
    print("🚀 Rolling Window Size Tuning (1M vs 2M vs 3M)")
    print("="*60)

    for sc_name, exog_col in scenarios.items():
        print(f"\n>> Scenario: {sc_name}")

        best_win_rmse = float('inf')
        best_win = -1

        for w in windows:
            try:
                # 모델 실행
                df_res = monthly_garch_midas_rolling(
                    df_10m, mfdfa_data, exog_col=exog_col,
                    midas_k=30,  # K=30 고정 (이미 검증됨)
                    window_months=w
                )

                if df_res.empty:
                    print(f"   - {w} Months: Failed (Empty Result)")
                    continue

                # RMSE 계산 준비
                flat_data, _ = prepare_flatten_features(df_res, mfdfa_data, lags=lags)

                # 실제값 vs 예측값
                actual = flat_data['Future_Vol_24h']
                pred = flat_data['Base_Vol'] * scale_factor

                rmse = np.sqrt(mean_squared_error(actual, pred))

                print(f"   - Window: {w} Month(s) | RMSE: {rmse:.4f}")

                results.append({
                    "Scenario": sc_name,
                    "Window Size": f"{w} Months",
                    "RMSE": rmse
                })

                if rmse < best_win_rmse:
                    best_win_rmse = rmse
                    best_win = w

            except Exception as e:
                print(f"   - {w} Months: Error ({e})")

        print(f" ✅ Best Window for {sc_name}: {best_win} Months (RMSE: {best_win_rmse:.4f})")

    # 결과표 출력
    print("\n" + "="*60)
    print("🏆 Window Size Tuning Leaderboard 🏆")
    print("="*60)
    res_df = pd.DataFrame(results).sort_values(by='RMSE')
    print(res_df.to_markdown(index=False))

    return res_df

# 실행
window_tuning_result = tune_rolling_window_size(df_10m_2021, mfdfa_2021)
```

```python
def run_daily_garch(df_daily, actual_rv_series, test_days):
    print("\n[Base Line] Running Daily GARCH(1, 1)")

    if 'return' in df_daily.columns:
        daily_ret = df_daily['return'].dropna()
    else:
        daily_ret = 100 * np.log(df_daily['close'] / df_daily['close'].shift(1)).dropna()

    #  인덱스가 날짜 형식이 아니면 변환
    if not isinstance(daily_ret.index, pd.DatetimeIndex):
        daily_ret.index = pd.to_datetime(daily_ret.index)

    # 타임존(Timezone) 강제 제거
    if daily_ret.index.tz is not None:
        daily_ret.index = daily_ret.index.tz_localize(None)

    model = arch_model(daily_ret, vol = 'Garch', p = 1, q = 1, rescale = False)
    res = model.fit(disp = 'off')

    actual_rv = actual_rv_series.resample('1D').last().dropna()

    test_start_date = actual_rv.index[-test_days]

    forecasts = res.forecast(start = test_start_date, horizon = 1)

    pred_vol_daily = np.sqrt(forecasts.variance['h.1'].dropna())

    common_idx = actual_rv.index.intersection(pred_vol_daily.index)

    mse = mean_squared_error(actual_rv.loc[common_idx], pred_vol_daily.loc[common_idx])
    rmse = np.sqrt(mse)

    print(f"-> Daily GARCH RMSE: {rmse:.4f}")

    return rmse

def run_intraday_garch(df_10m, actual_rv_series, test_days):
    print("\n[Base Line 2] Running Intraday GARCH(1, 1) - 10min Datas")

    df_temp = df_10m.copy()
    if df_temp.index.tz is not None:
        df_temp.index = df_temp.index.tz_localize(None)

    actual_rv_series = actual_rv_series.copy()
    if actual_rv_series.index.tz is not None:
        actual_rv_series.index = actual_rv_series.index.tz_localize(None)

    intra_ret = df_temp['return'].dropna()

    model = arch_model(intra_ret, vol = 'Garch', p = 1, q = 1, rescale = False)
    res = model.fit(disp = 'off', show_warning = False)

    test_start_idx = -test_days * 144

    cond_vol_10m = res.conditional_volatility.iloc[test_start_idx:]

    pred_vol_24h = cond_vol_10m * 12

    actual_rv = actual_rv_series.iloc[test_start_idx:]

    common_idx = actual_rv.index.intersection(pred_vol_24h.index)

    rmse = np.sqrt(mean_squared_error(actual_rv.loc[common_idx], pred_vol_24h.loc[common_idx]))

    print(f" -> Intraday GARCH RMSE: {rmse:.4f}")
    return rmse

def run_full_experiment(df_10m, mfdfa_data):
    results = []

    ai_models = ["LSTM", "Bi-LSTM", "GRU", "Bi-GRU", "Ridge"]

    garch_scenarios = {
        "Benchmark": None,
        "Delta_H": "delta_H",
        "Delta Alpha": "delta_alpha"
    }

    lags = 144
    TRAIN_WINDOW = 4320
    scale_factor = 12


    for g_name, exog_col in garch_scenarios.items():
        print(f"\n-> Scenario: {g_name} Processing")

        try:
            df_base_init = monthly_garch_midas(df_10m, mfdfa_data, exog_col = exog_col)
        except Exception as e:
            print(f"Error in {g_name}: {e}")
            continue

    print("\n[Data] Preparing Flatten Featrues")

    try:
        df_daily = pd.read_csv("data/BTCUSDT_1d_2020.csv")
        df_daily.columns = [c.lower().strip() for c in df_daily.columns]

        if 'datetime' in df_daily.columns:
            df_daily['timestamp'] = pd.to_datetime(df_daily['datetime'])
            df_daily = df_daily.set_index('timestamp').sort_index()
        elif 'date' in df_daily.columns:
            df_daily['date'] = pd.to_datetime(df_daily['date'])
            df_daily = df_daily.set_index('date').sort_index()

    except Exception as e:
        print(f"[Warning] Daily Data Load Failed: {e}")
        print(" -> df_10m을 리샘플링하여 대체합니다.")
        df_daily = df_10m.resample('1D').last().dropna()

    flat_data, features_all = prepare_flatten_features(df_base_init, mfdfa_data, lags = lags)
    TEST_DAYS = (len(flat_data) - TRAIN_WINDOW) // 144
    actual_rv_series = flat_data['Future_Vol_24h']

    rmse_daily = run_daily_garch(df_daily, actual_rv_series, TEST_DAYS)
    results.append({
        "Scenario": "1. Baseline (Daily)",
        "Model": "GARCH (Daily Data)",
        "Input Features": "Daily Close",
        "RMSE": rmse_daily,
        "Note": "Low frequency"
    })

    rmse_intra = run_intraday_garch(df_10m, actual_rv_series, TEST_DAYS)
    results.append({
        "Scenario": "2. Baseline (Intraday)",
        "Model": "GARCH (10m Data)",
        "Input Features": "10m Close",
        "RMSE": rmse_intra,
        "Note": "High frequency (Scaled)"
    })

    print(f"\nDirect AI Prediction (No GARCH)")
    try:
        df_direct = df_10m.copy().sort_index()
        if df_direct.index.tz is not None:
            df_direct.index = df_direct.index.tz_localize(None)
        df_direct = df_direct[~df_direct.index.duplicated(keep='first')]

        df_direct['Future_Vol_24h'] = (df_direct['return']**2).rolling(window=144).sum().shift(-144)
        df_direct['Future_Vol_24h'] = np.sqrt(df_direct['Future_Vol_24h'])

        df_direct['Past_RV'] = (df_direct['return']**2).rolling(window=144).sum()
        df_direct['Past_RV'] = np.sqrt(df_direct['Past_RV'])

        df_direct = df_direct.dropna()

        mfdfa_clean = mfdfa_data.copy()
        if mfdfa_clean.index.tz is not None:
            mfdfa_clean.index = mfdfa_clean.index.tz_localize(None)
        mfdfa_clean = mfdfa_clean[~mfdfa_clean.index.duplicated(keep='first')]

        df_direct['temp_date'] = df_direct.index.normalize()

        df_merged = df_direct.merge(
            mfdfa_clean,
            left_on='temp_date',
            right_index=True,
            how='left',
            validate='many_to_one'
        )

        df_merged = df_merged.fillna(method='ffill').dropna()
        if 'temp_date' in df_merged.columns:
            df_merged = df_merged.drop(columns=['temp_date'])

        print(f"   -> Direct Data Shape: {df_merged.shape}")

        # 4. Feature Set 정의
        feature_no_mf = ['Past_RV', 'return']
        mf_features = list(mfdfa_clean.columns)
        feature_full = feature_no_mf + mf_features

        direct_feature_sets = [
            ("No MF", feature_no_mf),
            ("Full (With MF)", feature_full)
        ]

        # 5. Direct AI Simulation Loop
        for ai_name in ai_models:
            for feat_name, feats in direct_feature_sets:
                try:
                    final_df = run_direct_rv_simulation(
                        df_merged, feats,
                        model_type=ai_name,
                        train_window=TRAIN_WINDOW
                    )

                    if not final_df.empty:
                        eval_df = final_df.dropna()
                        rmse_direct = np.sqrt(mean_squared_error(eval_df['Future_Vol_24h'], eval_df['Predicted_RV']))

                        results.append({
                            "Scenario": "Direct AI (No GARCH)",
                            "Model": f"Direct {ai_name}",
                            "Input Features": feat_name,
                            "RMSE": rmse_direct,
                            "Note": "Pure AI Prediction"
                        })
                        print(f"[Direct {ai_name} ({feat_name}) RMSE: {rmse_direct:.4f}]")

                except Exception as e:
                    print(f"Error in Direct {ai_name}: {e}")

    except Exception as e:
        print(f"[Error] Direct Phase Failed: {e}")


    for g_name, exog_col in garch_scenarios.items():
        print(f"\n-> Scenario: {g_name} Processing")

        try:
            df_base = monthly_garch_midas(df_10m, mfdfa_data, exog_col = exog_col)
        except Exception as e:
            print(f"Error in {g_name}: {e}")
            continue

        flat_data, features_all = prepare_flatten_features(df_base, mfdfa_data, lags = lags)
        garch_daily_pred = flat_data['Base_Vol'][-TEST_DAYS*144:] * scale_factor
        actual_rv = flat_data['Future_Vol_24h'][-TEST_DAYS*144:]

        rmse_gm = np.sqrt(mean_squared_error(actual_rv, garch_daily_pred))
        results.append({
            "Scenario": g_name,
            "Model": "GARCH-MIDAS",
            "Input Features": "10min Intraday",
            "RMSE": rmse_gm,
            "Note": "High frequency"
        })
        print(f"[{g_name} Base RMSE: {rmse_gm:.4f}")

        feature_no_mf = ['Past_RV', 'return']

        feature_full = features_all

        feature_sets = [
            ("No MF", feature_no_mf),
            ("Full (With MF)", feature_full)
        ]
        for ai_name in ai_models:
            for feat_type, feats in feature_sets:
                final_df, _ = run_rolling_simulation(
                    flat_data, feats,
                    model_type = ai_name,
                    train_window = TRAIN_WINDOW
                )

                eval_df = final_df.dropna()
                rmse_hybrid = np.sqrt(mean_squared_error(eval_df['Future_Vol_24h'], eval_df['Hybrid_Vol']))

                results.append({
                    "Scenario": g_name,
                    "Model": f"Hybrid {ai_name}",
                    "Input Features": feat_type,
                    "RMSE": rmse_hybrid,
                    "Note": "AI Correction"
                })

                print(f"[{g_name} + {ai_name} ({feat_type}) RMSE: {rmse_hybrid:.4f}]")

    results_df = pd.DataFrame(results)

    results_df = results_df.sort_values(by='RMSE', ascending = True)

    return results_df

print("="*80)
print("Final Ablation Study & Comparison")
print("="*80)

final_leaderboard = run_full_experiment(df_10m_2020, mfdfa_2020)

print(final_leaderboard.to_markdown(index=False))
```

```python
TUNING_SAVE_FILE = "Best_Tuning_Params_All.txt"

def tune_garch_midas_k(df_10m, mfdfa_data, exog_col, n_trials = 15):
    print(f"\n{'='*30} Tuning GARCH midas_k ({exog_col})\n{'='*30}")

    def objective(trial):
        k = trial.suggest_int('midas_k', 7, 30)

        try:
            df_base = monthly_garch_midas(df_10m, mfdfa_data, exog_col = exog_col, midas_k = k)
            flat_data, _ = prepare_flatten_features(df_base, mfdfa_data, lags = 144)

            TRAIN_WINDOW = 4320
            eval_df = flat_data.iloc[TRAIN_WINDOW:]

            rmse = np.sqrt(mean_squared_error(eval_df['Future_Vol_24h'], eval_df['Base_Vol'] * 12))

            return rmse
        except Exception:
            return float('inf')

    study = optuna.create_study(direction = 'minimize', sampler = TPESampler(seed = 42))
    study.optimize(objective, n_trials = n_trials)

    print(f"Best K: {study.best_params['midas_k']}, RMSE: {study.best_value:.4f}")
    return study.best_params['midas_k']


def tune_ai_hyperparameters(df_train, feature_cols, model_name, scenario_name, n_trials = 10):

  # 고유 키 값을 저장하여 각 모델을 구분 (LSTM (No GARCH) VS GARCH-MIDAS + LSTM)
  save_key = f"{scenario_name}_{model_name}"

  # 저장된 결과 확인 및 스킵 로직
  saved_results = {}
  if os.path.exists(TUNING_SAVE_FILE):
    with open(TUNING_SAVE_FILE, 'r', encoding = 'utf-8') as f:
      try: saved_results = json.load(f)
      except: pass

  if save_key in saved_results:
    print(f"\n [{save_key}] alerady tuned. Skipping")
    return saved_results[save_key]

  print(f"\n[{save_key}] Optuna Optimization Start")

  df_tune = df_train.copy()
  
  if 'Base_Vol' in df_tune.columns:
    df_tune['GARCH_Daily'] = df_tune['Base_Vol'] * 12
    df_tune['Residual'] = df_tune['Future_Vol_24h'] - df_tune['GARCH_Daily']
    y_full = df_tune[['Residual']].values
  else:
    y_full = df_tune[['Future_Vol_24h']].values
    
  X_full = df_tune[feature_cols].values

  TRAIN_WINDOW = 4320
  start_idx = TRAIN_WINDOW
  end_idx = len(df_train)

  scaler_X = MinMaxScaler((-1, 1)).fit(X_full[:start_idx])
  scaler_y = MinMaxScaler((-1, 1)).fit(y_full[:start_idx])

  X_scaled = scaler_X.transform(X_full)
  y_scaled = scaler_y.transform(y_full)

  future_vol_arr = df_tune['Future_Vol_24h'].values
  if 'Base_Vol' in df_tune.columns:
    garch_daily_arr = df_tune['GARCH_Daily'].values
    
  def objective(trial):
    params = {}
    if model_name in ['LSTM', 'BI-LSTM', 'GRU', 'BI-GRU']:
      params['hidden_dim'] = trial.suggest_categorical('hidden_dim', [32, 64, 128])
      params['num_layers'] = trial.suggest_int('num_layers', 1, 2)
      params['learning_rate'] = trial.suggest_float('learning_rate', 1e-4, 5e-3, log=True)
      # params['dropout'] = trial.suggest_float('dropout', 0.0, 0.5)
    elif model_name == 'Ridge':
      params['alpha'] = trial.suggest_float('alpha', 0.1, 10.0)

    # 추후 업데이트 빈도를 수정할 때 해당 값을 수정해서 각 빈도에 맞는 파라미터 튜닝
    UPDATE_FREQ = 1
    is_dl = model_name != 'Ridge'

    if is_dl:
      X_tensor = torch.FloatTensor(X_scaled).to(DEVICE).unsqueeze(1)
      y_tensor = torch.FloatTensor(y_scaled).to(DEVICE)
      input_dim = len(feature_cols)

      
      if model_name == 'LSTM':
        model = FlattenLSTM(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
      elif model_name == 'BI-LSTM':
        model = FlattenBiLSTM(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
      elif model_name == 'GRU':
        model = FlattenGRU(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
      elif model_name == 'BI-GRU':
        model = FlattenBiGRU(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)

      # if model_name == 'LSTM':
      #   model = FlattenLSTM_tuned(input_dim, params['hidden_dim'], params['num_layers'], params['dropout']).to(DEVICE)
      # elif model_name == 'BI-LSTM':
      #   model = FlattenBiLSTM_tuned(input_dim, params['hidden_dim'], params['num_layers'], params['dropout']).to(DEVICE)
      # elif model_name == 'GRU':
      #   model = FlattenGRU_tuned(input_dim, params['hidden_dim'], params['num_layers'], params['dropout']).to(DEVICE)
      # elif model_name == 'BI-GRU':
      #   model = FlattenBiGRU_tuned(input_dim, params['hidden_dim'], params['num_layers'], params['dropout']).to(DEVICE)

      optimizer = torch.optim.Adam(model.parameters(), lr = params['learning_rate'])
      loss_fn = nn.HuberLoss(delta = 1.0).to(DEVICE)
    else:
      ridge_model = Ridge(alpha = params['alpha'])

    predictions, actuals = [], []

    pbar = tqdm(range(start_idx, end_idx), desc = f"{model_name} Tuning")

    for t in pbar:
      if (t - start_idx) % UPDATE_FREQ == 0:
        if is_dl:
          model.train()
          optimizer.zero_grad()
          pred = model(X_tensor[t - TRAIN_WINDOW : t])
          loss = loss_fn(pred, y_tensor[t - TRAIN_WINDOW : t])
          loss.backward()
          optimizer.step()
        else:
          ridge_model.fit(X_scaled[t - TRAIN_WINDOW : t], y_scaled[t - TRAIN_WINDOW : t])

      if is_dl:
        model.eval()
        with torch.no_grad():
          pred_val = model(X_tensor[t].unsqueeze(0)).item()
      else:
        pred_val = ridge_model.predict(X_scaled[t].reshape(1, -1)).item()

      predictions.append(pred_val)
      actuals.append(y_scaled[t][0])

      step_count = t - start_idx + 1
      if step_count > 0 and step_count % 4320 == 0:
        temp_pred_arr = np.array(predictions).reshape(-1, 1)
        temp_pred_rescaled = scaler_y.inverse_transform(temp_pred_arr).flatten()
        temp_actual_future = future_vol_arr[start_idx : start_idx + step_count]

        if 'Base_Vol' in df_tune.columns:
            # Hybrid: GARCH_Daily + 역변환된 예측 잔차
            temp_garch = garch_daily_arr[start_idx : start_idx + step_count]
            temp_final_pred = temp_garch + temp_pred_rescaled
        else:
            # Direct
            temp_final_pred = temp_pred_rescaled
        temp_rmse = np.sqrt(mean_squared_error(temp_actual_future, temp_final_pred))
        trial.report(temp_rmse, step_count // 4320)
        if trial.should_prune():
          raise optuna.TrialPruned()
    pred_array = np.array(predictions).reshape(-1, 1)
    pred_rescaled = scaler_y.inverse_transform(pred_array).flatten()
    actual_future = future_vol_arr[start_idx:end_idx]
    
    if 'Base_Vol' in df_tune.columns:
      final_pred_vol = garch_daily_arr[start_idx:end_idx] + pred_rescaled
    else:
      final_pred_vol = pred_rescaled
      
    rmse = np.sqrt(mean_squared_error(actual_future, final_pred_vol))

    return rmse

  pruner = optuna.pruners.MedianPruner(n_startup_trials = 5, n_warmup_steps = 2)
  study = optuna.create_study(direction = 'minimize', pruner = pruner)
  study.optimize(objective, n_trials = n_trials)

  saved_results[save_key] = study.best_params
  with open(TUNING_SAVE_FILE, 'w', encoding = 'utf-8') as f:
    json.dump(saved_results, f, indent = 4)

  print(f"[{save_key}] Best Performance: {study.best_value:.4f}")
  return study.best_params
```

```python
DAILY_TUNING_SAVE_FILE = "Daily_Direct_Tuning_Params_2020.txt"

def prepare_daily_direct_data(df_10m, df_daily):
    print("[INFO] Preparing Daily Data for Direct AI")
    df_new = df_daily.copy()
    
    if not isinstance(df_new.index, pd.DatetimeIndex):
        cols_lower = {c.lower(): c for c in df_new.columns}
        for time_col in ['timestamp', 'date', 'datetime', 'time']:
            if time_col in cols_lower:
                actual_col_name = cols_lower[time_col]
                df_new.index = pd.to_datetime(df_new[actual_col_name])
                break
    
    daily_rv = (df_10m['return']**2).resample('1D').sum().apply(np.sqrt)
    
    if hasattr(daily_rv.index, 'tz') and daily_rv.index.tz is not None:
        daily_rv.index = daily_rv.index.tz_localize(None)
    if hasattr(df_new.index, 'tz') and df_new.index.tz is not None:
        df_new.index = df_new.index.tz_localize(None)
        
    daily_rv.index = daily_rv.index.normalize()
    df_new.index = df_new.index.normalize()
    
    df_new['Past_RV'] = daily_rv
    df_new['Future_Vol_1d'] = daily_rv.shift(-1)
    
    if 'close' in df_new.columns:
        df_new['daily_return'] = df_new['close'].pct_change()
    else:
        df_new['daily_return'] = 0
        
    df_new = df_new.dropna()
    print(f"-> Daily Data Shape: {df_new.shape}")
    return df_new

def tune_daily_direct_ai(df_train, feature_cols, model_name, n_trials = 10):
    save_key = f"Daily_Direct_{model_name}"
    
    saved_results = {}
    if os.path.exists(DAILY_TUNING_SAVE_FILE):
        with open(DAILY_TUNING_SAVE_FILE, 'r', encoding = 'utf-8') as f:
            try: saved_results = json.load(f)
            except: pass
    
    if save_key in saved_results:
        print(f"\n[{save_key}] already tuned. Skipping")
        return saved_results[save_key]
    
    print(f"\n[{save_key}] Optuna Optimizer Start")
    
    X_full = df_train[feature_cols].values
    y_full = df_train[['Future_Vol_1d']].values
    
    TRAIN_WINDOW = 32
    start_idx = TRAIN_WINDOW
    end_idx = len(df_train)
    
    scaler_X = MinMaxScaler((-1, 1)).fit(X_full[:start_idx])
    scaler_y = MinMaxScaler((-1, 1)).fit(y_full[:start_idx])
    
    X_scaled = scaler_X.transform(X_full)
    y_scaled = scaler_y.transform(y_full)
    
    actual_future_vols = df_train['Future_Vol_1d'].values
    
    def objective(trial):
        params = {}
        if model_name in ['LSTM', 'BI-LSTM', 'GRU', 'BI-GRU']:
            params['hidden_dim'] = trial.suggest_categorical('hidden_dim', [32, 64, 128])
            params['num_layers'] = trial.suggest_int('num_layers', 1, 2)
            params['learning_rate'] = trial.suggest_float('learning_rate', 1e-4, 5e-3, log = True)
        elif model_name == 'Ridge':
            params['alpha'] = trial.suggest_float('alpha', 0.1, 10.0)
            
        UPDATE_FREQ = 1
        is_dl = model_name != 'Ridge'
        
        if is_dl:
            X_tensor = torch.FloatTensor(X_scaled).to(DEVICE).unsqueeze(1)
            y_tensor = torch.FloatTensor(y_scaled).to(DEVICE)
            input_dim = len(feature_cols)
            
            if model_name == 'LSTM':
                model = FlattenLSTM(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
            elif model_name == 'BI-LSTM':
                model = FlattenBiLSTM(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
            elif model_name == 'GRU':
                model = FlattenGRU(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
            elif model_name == 'BI-GRU':
                model = FlattenBiGRU(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
                
            optimizer = torch.optim.Adam(model.parameters(), lr = params['learning_rate'])
            loss_fn = nn.HuberLoss(delta = 1.0).to(DEVICE)
        else:
            ridge_model = Ridge(alpha = params['alpha'])
            
        predictions = []
        
        for t in range(start_idx, end_idx):
            if (t - start_idx) % UPDATE_FREQ == 0:
                if is_dl:
                    model.train()
                    optimizer.zero_grad()
                    pred = model(X_tensor[t - TRAIN_WINDOW : t])
                    loss = loss_fn(pred, y_tensor[t - TRAIN_WINDOW : t])
                    loss.backward()
                    optimizer.step()
                else:
                    ridge_model.fit(X_scaled[t - TRAIN_WINDOW : t], y_scaled[t - TRAIN_WINDOW : t])
                    
            if is_dl:
                model.eval()
                with torch.no_grad():
                    pred_val = model(X_tensor[t].unsqueeze(0)).item()
            else:
                pred_val = ridge_model.predict(X_scaled[t].reshape(1, -1)).item()
                
            predictions.append(pred_val)
            
        pred_array = np.array(predictions).reshape(-1, 1)
        pred_rescaled = scaler_y.inverse_transform(pred_array).flatten()
        actual_future = actual_future_vols[start_idx : end_idx]
        
        rmse = np.sqrt(mean_squared_error(actual_future, pred_rescaled))
        return rmse
    
    pruner = optuna.pruners.MedianPruner(n_startup_trials = 5, n_warmup_steps = 2)
    study = optuna.create_study(direction = 'minimize', pruner = pruner)
    study.optimize(objective, n_trials = n_trials)
    
    saved_results[save_key] = study.best_params
    with open(DAILY_TUNING_SAVE_FILE, 'w', encoding = 'utf-8') as f:
        json.dump(saved_results, f, indent = 4)
        
    print(f"[{save_key}] Best Performance (Daily): {study.best_value:.4f}")
    return study.best_params

if __name__ == "__main__":
    df_daily = pd.read_csv("data/BTCUSDT_1d_2020.csv")
    
    df_daily.columns = [c.lower().strip() for c in df_daily.columns]
    
    daily_direct_data = prepare_daily_direct_data(df_10m_2020, df_daily)
    
    daily_features = ['Past_RV', 'daily_return']
    
    daily_ai_models = ["LSTM", "BI-LSTM", "GRU", "BI-GRU", "Ridge"]
    
    print("\n" + "="*30)
    print("Starting Daily Direct AI Tuning")
    print("="*30)
    
    for ai_name in daily_ai_models:
        best_daily_p = tune_daily_direct_ai(
            daily_direct_data,
            daily_features,
            ai_name,
            n_trials = 30
        )
```

```python
def run_full_experiment_tuned(df_10m, mfdfa_data):

    final_results = []

    midas_k = 14

    ridge_params = {'alpha': [0.1, 1.0, 5.0, 10.0]}

    dl_params = {
        'hidden_dim': [32, 64, 128],
        'num_layers': [1, 2],
        'learning_rate': [0.01, 0.001],
        'dropout': [0.0, 0.2]
    }

    ai_models = ["Ridge", "LSTM", "Bi-LSTM", "GRU", "Bi-GRU"]

    scenarios = {
        "Benchmark": None,
        "Delta_H": "delta_H",
        "Delta_Alpha": "delta_alpha"
    }

    lags = 144
    TRAIN_WINDOW = 4320
    scale_factor = 12

    monthly_vol = df_10m['return'].resample('MS').std()
    valid_vol = monthly_vol.iloc[2:]
    target_month = valid_vol.idxmax()

    if target_month.tzinfo is not None:
        target_month = target_month.tz_localize(None)

    target_start_ts = target_month
    target_end_ts = target_month + pd.DateOffset(months=1)

    print(f"   -> Target Month: {target_start_ts.strftime('%Y-%m')} (Highest Volatility)")

    for sc_name, exog_col in scenarios.items():
        print(f"\n{'='*60}\nScenario: {sc_name}\n{'='*60}")

        best_g_rmse = float('inf')
        best_df_base = None

        print("Tuning MIDAS K")
        try:
            temp_base = monthly_garch_midas_rolling_tuned(df_10m, mfdfa_data, exog_col = exog_col, midas_k = midas_k)
            flat, _ = prepare_flatten_features(temp_base,mfdfa_data, lags = lags)

            t_start = flat.index.searchsorted(target_start_ts)
            t_end = flat.index.searchsorted(target_end_ts)

            eval_df = flat.iloc[t_start:t_end:]

            rmse = np.sqrt(mean_squared_error(eval_df['Future_Vol_24h'], eval_df['Base_Vol'] * scale_factor))
            print(f"  -K = {midas_k} RMSE: {rmse:.4f}")

            if rmse < best_g_rmse:
                best_g_rmse = rmse
                best_df_base = temp_base
        except Exception as e:
                print(f"  -K={midas_k} Failed: {e}")

        print(f" Best K for {sc_name}: {midas_k} (RMSE: {best_g_rmse:.4f})")

        final_results.append({
            "Scenario": sc_name, "Model": "GARCH (Base)",
            "Params": f"K={midas_k}", "RMSE": best_g_rmse
        })

        flat_data, features_all = prepare_flatten_features(best_df_base, mfdfa_data, lags = lags)

        stress_start_idx = flat_data.index.get_loc(target_start_ts)
        stress_end_idx = flat_data.index.get_loc(target_end_ts)
        custom_range = (stress_start_idx, stress_end_idx)

        for ai_name in ai_models:
            print(f"\n >> {ai_name} Tuning")

            grid = ParameterGrid(ridge_params if ai_name == 'Ridge' else dl_params)
            model_history = []

            pbar = tqdm(grid, desc=f"    {ai_name} Grid", leave=True)

            for params in pbar:
                try:
                    final_df, _ = run_rolling_simulation_tuned(
                        flat_data, features_all,
                        model_type = ai_name,
                        train_window = TRAIN_WINDOW,
                        params = params,
                        custom_range = custom_range
                    )

                    eval_df = final_df.dropna()
                    current_rmse = np.sqrt(mean_squared_error(eval_df['Future_Vol_24h'], eval_df['Hybrid_Vol']))

                    model_history.append({
                    "Params": str(params),
                    "RMSE": current_rmse
                    })
                    pbar.set_postfix({'RMSE': f"{current_rmse:.4f}"})

                except Exception as e:
                    print(f" Final Run Failed: {e}")
                gc.collect()

            if len(model_history) > 0:
                df_history = pd.DataFrame(model_history).sort_values(by = 'RMSE')
                top3 = df_history.head(3)

                for rank, row in enumerate(top3.itertuples(), 1):
                    final_results.append({
                        "Scenario": sc_name,
                        "Model": f"Hybrid {ai_name}",
                        "Rank": rank, # 등수 추가
                        "Params": row.Params,
                        "RMSE": row.RMSE
                    })
            else:
                print(F" -> {ai_name} Failed to produce results")

    return pd.DataFrame(final_results).sort_values(by=['Scenario', 'Model', 'Rank'])
# final_leaderboard_tuned = run_full_experiment_tuned(df_10m, mfdfa_2021)

# print("\n" + "="*80)
# print(" Final Optimized Benchmark Leaderboard ")
# print("="*80)

# try:
#     print(final_leaderboard_tuned.to_markdown(index=False))
# except:
#     print(final_leaderboard_tuned)
```

```python
def run_daily_garch(df_daily, actual_rv_series, test_days):
    print("\n[Base Line] Running Daily GARCH(1, 1)")

    if 'return' in df_daily.columns:
        daily_ret = df_daily['return'].dropna()
    else:
        daily_ret = 100 * np.log(df_daily['close'] / df_daily['close'].shift(1)).dropna()

    #  인덱스가 날짜 형식이 아니면 변환
    if not isinstance(daily_ret.index, pd.DatetimeIndex):
        daily_ret.index = pd.to_datetime(daily_ret.index)

    # 타임존(Timezone) 강제 제거
    if daily_ret.index.tz is not None:
        daily_ret.index = daily_ret.index.tz_localize(None)

    model = arch_model(daily_ret, vol = 'Garch', p = 1, q = 1, rescale = False)
    res = model.fit(disp = 'off')

    actual_rv = actual_rv_series.resample('1D').last().dropna()

    test_start_date = actual_rv.index[-test_days]

    forecasts = res.forecast(start = test_start_date, horizon = 1)

    pred_vol_daily = np.sqrt(forecasts.variance['h.1'].dropna())

    common_idx = actual_rv.index.intersection(pred_vol_daily.index)

    mse = mean_squared_error(actual_rv.loc[common_idx], pred_vol_daily.loc[common_idx])
    rmse = np.sqrt(mse)

    print(f"-> Daily GARCH RMSE: {rmse:.4f}")

    return rmse

def run_intraday_garch(df_10m, actual_rv_series, test_days):
    print("\n[Base Line 2] Running Intraday GARCH(1, 1) - 10min Datas")

    df_temp = df_10m.copy()
    if df_temp.index.tz is not None:
        df_temp.index = df_temp.index.tz_localize(None)

    actual_rv_series = actual_rv_series.copy()
    if actual_rv_series.index.tz is not None:
        actual_rv_series.index = actual_rv_series.index.tz_localize(None)

    intra_ret = df_temp['return'].dropna()

    model = arch_model(intra_ret, vol = 'Garch', p = 1, q = 1, rescale = False)
    res = model.fit(disp = 'off', show_warning = False)

    test_start_idx = -test_days * 144

    cond_vol_10m = res.conditional_volatility.iloc[test_start_idx:]

    pred_vol_24h = cond_vol_10m * 12

    actual_rv = actual_rv_series.iloc[test_start_idx:]

    common_idx = actual_rv.index.intersection(pred_vol_24h.index)

    rmse = np.sqrt(mean_squared_error(actual_rv.loc[common_idx], pred_vol_24h.loc[common_idx]))

    print(f" -> Intraday GARCH RMSE: {rmse:.4f}")
    return rmse

def run_full_experiment(df_10m, mfdfa_data):
    results = []

    DEFAULT_PARAMS = {
        "Ridge":   {'alpha': 1.0},          # 기본값
        "LSTM":    {'hidden_dim': 64, 'num_layers': 1, 'learning_rate': 0.001, 'dropout': 0.0},
        "Bi-LSTM": {'hidden_dim': 64, 'num_layers': 1, 'learning_rate': 0.001, 'dropout': 0.0},
        "GRU":     {'hidden_dim': 64, 'num_layers': 1, 'learning_rate': 0.001, 'dropout': 0.0},
        "Bi-GRU":  {'hidden_dim': 64, 'num_layers': 1, 'learning_rate': 0.001, 'dropout': 0.0}
    }

    ai_models = ["LSTM", "Bi-LSTM", "GRU", "Bi-GRU", "Ridge"]

    garch_scenarios = {
        "Benchmark": None,
        "Delta_H": "delta_H",
        "Delta Alpha": "delta_alpha"
    }

    lags = 144
    TRAIN_WINDOW = 4320
    scale_factor = 12
    FIXED_K = 14


    print("\n[Data] Preparing Flatten Featrues")

    try:
        df_daily = pd.read_csv("data/BTCUSDT_1d_2021.csv")
        df_daily.columns = [c.lower().strip() for c in df_daily.columns]

        if 'datetime' in df_daily.columns:
            df_daily['timestamp'] = pd.to_datetime(df_daily['datetime'])
            df_daily = df_daily.set_index('timestamp').sort_index()
        elif 'date' in df_daily.columns:
            df_daily['date'] = pd.to_datetime(df_daily['date'])
            df_daily = df_daily.set_index('date').sort_index()

        temp_base = monthly_garch_midas(df_10m, mfdfa_data, exog_col = None, midas_k = FIXED_K)

        flat_data, _ = prepare_flatten_features(temp_base, mfdfa_data, lags = lags)
        actual_rv_series = flat_data['Future_Vol_24h']

        TEST_DAYS = (len(flat_data) - TRAIN_WINDOW) // 144

        rmse_daily = run_daily_garch(df_daily, actual_rv_series, TEST_DAYS)
        results.append({
            "Scenario": "1. Baseline (Daily)",
            "Model": "GARCH (Daily Data)",
            "Input Features": "Daily Close",
            "RMSE": rmse_daily,
            "Note": "Low frequency"
        })

        rmse_intra = run_intraday_garch(df_10m, actual_rv_series, TEST_DAYS)
        results.append({
            "Scenario": "2. Baseline (Intraday)",
            "Model": "GARCH (10m Data)",
            "Input Features": "10m Close",
            "RMSE": rmse_intra,
            "Note": "High frequency (Scaled)"
        })

    except Exception as e:
        print(f"[Warning] Daily Data Load Failed: {e}")

    print(f"\n{'='*20} Direct AI Prediction {'='*20}")
    try:
        df_direct = df_10m.copy()

        if df_direct.index.tz is not None:
            df_direct.index = df_direct.index.tz_localize(None)

        df_direct = df_direct[~df_direct.index.duplicated(keep='first')]

        df_direct['Rolling_Vol'] = df_direct['return'].rolling(window=144).std() * np.sqrt(144)
        df_direct['Future_Vol_24h'] = df_direct['Rolling_Vol'].shift(-144)

        df_direct['Past_RV'] = df_direct['return'].rolling(window = 144).std() * np.sqrt(144)

        mfdfa_clean = mfdfa_data.copy()
        if mfdfa_clean.index.tz is not None:
            mfdfa_clean.index = mfdfa_clean.index.tz_localize(None)

        mfdfa_clean = mfdfa_clean[~mfdfa_clean.index.duplicated(keep='first')]

        df_direct['temp_date'] = df_direct.index.normalize()

        df_merged = df_direct.merge(mfdfa_clean, left_on='temp_date', right_index=True, how='left', validate='m:1')

        df_merged = df_merged.fillna(method = 'ffill').dropna()

        if 'temp_date' in df_merged.columns:
            df_merged = df_merged.drop(columns = ['temp_date'])

        feature_no_mf = ['Past_RV', 'return']

        mf_features = list(mfdfa_clean.columns)
        feature_full = feature_no_mf + mf_features

        direct_feature_sets = [
            ("No MF", feature_no_mf),
            ("Full (With MF)", feature_full)
        ]

        for ai_name in ai_models:
            params = DEFAULT_PARAMS[ai_name]
            for feat_name, feats in direct_feature_sets:
                try:
                    final_df = run_direct_rv_simulation(
                        df_merged, feats,
                        model_type = ai_name,
                        train_window = TRAIN_WINDOW,
                        params = params
                    )

                    if not final_df.empty:
                        eval_df = final_df.copy()
                        rmse_direct = np.sqrt(mean_squared_error(eval_df['Future_Vol_24h'], eval_df['Predicted_RV']))

                        results.append({
                            "Scenario": "Direct AI (No GARCH)",
                            "Model": f"Direct {ai_name}",
                            "Input Features": feat_name,
                            "RMSE": rmse_direct,
                            "Note": "Pure AI Prediction"
                        })
                        print(f"Direct {ai_name} ({feat_name}) RMSE: {rmse_direct:.4f}")
                except Exception as e:
                    print(f"Error in Direct {ai_name}: {e}")
    except Exception as e:
        print(f"Direct Phase Failed: {e}")

    for g_name, exog_col in garch_scenarios.items():
        print(f"\n-> Scenario: {g_name} Processing")

        try:
            df_base = monthly_garch_midas(df_10m, mfdfa_data, exog_col = exog_col, midas_k = FIXED_K)

            flat_data, features_all = prepare_flatten_features(df_base, mfdfa_data, lags = lags)

            eval_df_garch = flat_data.iloc[TRAIN_WINDOW:]
            garch_daily_pred = eval_df_garch['Base_Vol'] * scale_factor
            actual_rv = eval_df_garch['Future_Vol_24h']

            rmse_gm = np.sqrt(mean_squared_error(actual_rv, garch_daily_pred))
            results.append({
                "Scenario": g_name,
                "Model": "GARCH-MIDAS",
                "Input Features": "10min Intraday",
                "RMSE": rmse_gm,
                "Note": "High frequency"
            })
            print(f"[{g_name} Base RMSE: {rmse_gm:.4f}")

            feature_no_mf = ['Past_RV', 'return']

            feature_full = features_all

            feature_sets = [
                ("No MF", feature_no_mf),
                ("Full (With MF)", feature_full)
            ]
            for ai_name in ai_models:
                params = DEFAULT_PARAMS[ai_name]
                for feat_type, feats in feature_sets:
                    final_df, _ = run_rolling_simulation_tuned(
                        flat_data, feats,
                        model_type = ai_name,
                        train_window = TRAIN_WINDOW,
                        params = params,
                        custom_range = None
                    )

                    eval_df = final_df.dropna()
                    rmse_hybrid = np.sqrt(mean_squared_error(eval_df['Future_Vol_24h'], eval_df['Hybrid_Vol']))

                    results.append({
                        "Scenario": g_name,
                        "Model": f"Hybrid {ai_name}",
                        "Input Features": feat_type,
                        "RMSE": rmse_hybrid,
                        "Note": "AI Correction"
                    })

                    print(f"[{g_name} + {ai_name} ({feat_type}) RMSE: {rmse_hybrid:.4f}]")
        except Exception as e:
            print(f"Error in {g_name}: {e}")
            continue


    results_df = pd.DataFrame(results)

    results_df = results_df.sort_values(by='RMSE', ascending = True)

    return results_df

print("="*80)
print("Final Ablation Study & Comparison")
print("="*80)

final_leaderboard = run_full_experiment(df_10m, mfdfa_2021)

print(final_leaderboard.to_markdown(index=False))
```

```python
# 10분봉 데이터 업데이트 빈도 (1 = 10분 마다, 6 = 1시간 마다, 144 = 1일 마다)
INTRADAY_UPDATE_FREQ = 1

# 일별 데이터 업데이트 주기
DAILY_UPDATE_FREQ = 1

FINAL_RESULT_FILE = ""
PREDICTIONS_CSV_FILE = ""

TUNING_10M_FILE = "Best_Tuning_Params_All.txt"
TUNING_DAILY_FILE = "Daily_Direct_Tuning_Params.txt"

def log_result(scenario, model, feature_set, is_tuned, rmse, note):
    tune_status = "Tuned" if is_tuned else "Untuned"
    result_str = f"[{scenario}] | Model: {model} ({tune_status}) | Feat: {feature_set} | RMSE: {rmse:.4f} | Note: {note}\n"
    print(f"{result_str.strip()}")
    
    with open(FINAL_RESULT_FILE, 'a', encoding='utf-8') as f:
        f.write(result_str)

def save_predictions_to_csv(dates, actuals, preds, scenario, model_name, feature_name, is_tuned):
    tune_status = "Tuned" if is_tuned else "Untuned"
    full_model_name = f"[{scenario}] {model_name} ({tune_status})"
    
    df_log = pd.DataFrame({
        'Date': dates,
        'Actual_RV': actuals,
        'Predicted_RV': preds,
        'Model_Name': full_model_name,
        'Feature': feature_name
    })
    
    file_exists = os.path.isfile(PREDICTIONS_CSV_FILE)
    df_log.to_csv(PREDICTIONS_CSV_FILE, mode = 'a', header = not file_exists, index = False)
   
def run_universal_ai_simulation(df, feature_cols, target_col, model_name, params, train_window, update_freq):
    X_full = df[feature_cols].values
    y_full = df[[target_col]].values
    
    start_idx = train_window
    end_idx = len(df)
    
    scaler_X = MinMaxScaler((-1, 1)).fit(X_full[:start_idx])
    scaler_y = MinMaxScaler((-1, 1)).fit(y_full[:start_idx])
    
    X_scaled = scaler_X.transform(X_full)
    y_scaled = scaler_y.transform(y_full)
    
    if params is None:
        if model_name != 'Ridge':
            params = {'hidden_dim': 64, 'num_layers': 2, 'learning_rate': 0.001}
        else:
            params = {'alpha': 1.0}
            
    is_dl = model_name != 'Ridge'
    
    if is_dl:
        X_tensor = torch.FloatTensor(X_scaled).to(DEVICE).unsqueeze(1)
        y_tensor = torch.FloatTensor(y_scaled).to(DEVICE)
        input_dim = len(feature_cols)
        
        if model_name == 'LSTM': model = FlattenLSTM(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
        elif model_name == 'BI-LSTM': model = FlattenBiLSTM(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
        elif model_name == 'GRU': model = FlattenGRU(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
        elif model_name == 'BI-GRU': model = FlattenBiGRU(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
            
        optimizer = torch.optim.Adam(model.parameters(), lr=params['learning_rate'])
        loss_fn = nn.HuberLoss(delta = 1.0).to(DEVICE)
    else:
        ridge_model = Ridge(alpha=params['alpha'])

    predictions = []
    
    pbar = tqdm(range(start_idx, end_idx), desc=f"Evaluating {model_name}", leave=False)
    for t in pbar:
        if (t - start_idx) % update_freq == 0:
            if is_dl:
                model.train()
                optimizer.zero_grad()
                pred = model(X_tensor[t - train_window : t])
                loss = loss_fn(pred, y_tensor[t - train_window : t])
                loss.backward()
                optimizer.step()
            else:
                ridge_model.fit(X_scaled[t - train_window : t], y_scaled[t - train_window : t])
                
        if is_dl:
            model.eval()
            with torch.no_grad():
                pred_val = model(X_tensor[t].unsqueeze(0)).item()
        else:
            pred_val = ridge_model.predict(X_scaled[t].reshape(1, -1)).item()
            
        predictions.append(pred_val)
        
    pred_array = np.array(predictions).reshape(-1, 1)
    pred_rescaled = scaler_y.inverse_transform(pred_array).flatten()
    
    return pred_rescaled

def run_final_evaluation(df_10m, df_daily, mfdfa_data, current_year):
    print("\n" + "="*70)
    print(f"FINAL EVALUATION SUITE (Year: {current_year} / Update Freq: {INTRADAY_UPDATE_FREQ})")
    print(f"Saving to: {FINAL_RESULT_FILE}")
    print("="*70)
    
    if not os.path.exists(FINAL_RESULT_FILE):
        with open(FINAL_RESULT_FILE, 'w', encoding='utf-8') as f:
            f.write(f"--- Evaluation for Year: {current_year} | Freq: {INTRADAY_UPDATE_FREQ} ---\n")
            f.write("Scenario | Model | Feature Set | RMSE | Note\n")
            f.write("-" * 80 + "\n")
    
    completed_evals = set()
    with open(FINAL_RESULT_FILE, 'r', encoding = 'utf-8') as f:
        for line in f:
            if "|" in line and "RMSE" in line:
                completed_evals.add(line.strip())

    garch_scenarios = {
        "Benchmark": None,
        "Delta_H": "delta_H",
        "Delta_Alpha": "delta_alpha"
    }
    
    ai_models = ["LSTM", "BI-LSTM", "GRU", "BI-GRU", "Ridge"]
    
    tuned_10m = {}
    tuned_daily = {}
    try:
        with open(TUNING_10M_FILE, 'r') as f: tuned_10m = json.load(f)
        with open(TUNING_DAILY_FILE, 'r') as f: tuned_daily = json.load(f)
    except Exception as e:
        print(f"[Warning] Tuning file load failed: {e}. Tuned runs will be skipped.")

    print(f"\n[Phase 1] Evaluating Baseline GARCH Models ({current_year})")
    
    df_temp = monthly_garch_midas(df_10m, mfdfa_data, exog_col=None)
    flat_temp, _ = prepare_flatten_features(df_temp, mfdfa_data, lags=144)
    TEST_DAYS_10M = (len(flat_temp) - 4320) // 144
    actual_rv_series = flat_temp['Future_Vol_24h']
    
    # 1. Daily & Intraday 기본 GARCH
    if not any("[1. Baseline]" in line for line in completed_evals):
        rmse_daily = run_daily_garch(df_daily, actual_rv_series, TEST_DAYS_10M)
        log_result("1. Baseline", "GARCH(1,1)", "Daily Return", False, rmse_daily, "Low Freq")
    else:
        print("[Skip] [1. Baseline] GARCH(1, 1)")
        
    if not any("[2. Baseline]" in line for line in completed_evals):
        rmse_intra = run_intraday_garch(df_10m, actual_rv_series, TEST_DAYS_10M)
        log_result("2. Baseline", "GARCH(1,1)", "10m Return", False, rmse_intra, "High Freq")
    else:
        print("[Skip] [2. Baseline] GARCH(1, 1)")
    
    # 2. GARCH-MIDAS 시나리오별 평가 (Benchmark, Delta_H, Delta_Alpha)
    for g_name, exog_col in garch_scenarios.items():
        eval_id = f"[3. Baseline GM] | Model: GARCH-MIDAS ({g_name})"
        if any(eval_id in line for line in completed_evals):
            print(f"[Skip] {eval_id}")
            continue
        
        try:
            df_base = monthly_garch_midas(df_10m, mfdfa_data, exog_col=exog_col)
            flat_data, _ = prepare_flatten_features(df_base, mfdfa_data, lags=144)
            garch_midas_pred = flat_data['Base_Vol'][-TEST_DAYS_10M*144:] * 12
            actual_rv_test = flat_data['Future_Vol_24h'][-TEST_DAYS_10M*144:]
            
            rmse_gm = np.sqrt(mean_squared_error(actual_rv_test, garch_midas_pred))
            log_result(f"3. Baseline GM", f"GARCH-MIDAS ({g_name})", f"10m + {exog_col}", False, rmse_gm, "Macro")
        except Exception as e:
            print(f"[Error] Baseline GARCH-MIDAS ({g_name}) Failed: {e}")


    # ---------------------------------------------------------
    # [Phase 2] Direct AI (Daily) 평가
    # ---------------------------------------------------------
    print(f"\n▶ [Phase 2] Evaluating Direct AI (Daily Data - {current_year})")
    daily_data = prepare_daily_direct_data(df_10m, df_daily)
    daily_feats = ['Past_RV', 'daily_return']
    target_daily = 'Future_Vol_1d'
    actual_daily_test = daily_data[target_daily].values[30:]
    test_dates_daily = daily_data.index[30:]
    
    for ai_name in ai_models:
        for is_tuned in [False, True]:
            tune_status = "Tuned" if is_tuned else "Untuned"
            eval_id = f"[Direct AI (Daily)] | Model: {ai_name} ({tune_status}) | Feat: Daily Feat"
            
            if any(eval_id in line for line in completed_evals):
                print(f"[Skip] {eval_id} (Already calculated)")
                continue
            
            params = tuned_daily.get(f"Daily_Direct_{ai_name}", None) if is_tuned else None
            if is_tuned and params is None: continue 
                
            preds = run_universal_ai_simulation(
                daily_data, daily_feats, target_daily, ai_name, params, train_window=30, update_freq=DAILY_UPDATE_FREQ
            )
            rmse = np.sqrt(mean_squared_error(actual_daily_test, preds))
            log_result("Direct AI (Daily)", ai_name, "Daily Feat", is_tuned, rmse, "Low Freq AI")
            save_predictions_to_csv(test_dates_daily, actual_daily_test, preds, "Direct AI (Daily)", ai_name, "Daily Feat", is_tuned)

    # ---------------------------------------------------------
    # [Phase 3] Direct AI (10min) 평가
    # ---------------------------------------------------------
    print(f"\n▶ [Phase 3] Evaluating Direct AI (10min Data - {current_year})")
    df_direct = df_10m.copy()
    df_direct.index = df_direct.index.tz_localize(None)
    df_direct = df_direct[~df_direct.index.duplicated(keep='first')]
    df_direct['Future_Vol_24h'] = (df_direct['return']**2).rolling(window=144).sum().shift(-144).apply(np.sqrt)
    df_direct['Past_RV'] = (df_direct['return']**2).rolling(window=144).sum().apply(np.sqrt)
    df_direct = df_direct.dropna()
    
    mfdfa_clean = mfdfa_data.copy()
    mfdfa_clean.index = mfdfa_clean.index.tz_localize(None)
    mfdfa_clean = mfdfa_clean[~mfdfa_clean.index.duplicated(keep='first')]
    
    df_direct['temp_date'] = df_direct.index.normalize()
    direct_merged = df_direct.merge(mfdfa_clean, left_on='temp_date', right_index=True, how='left')
    direct_merged = direct_merged.fillna(method='ffill').dropna().drop(columns=['temp_date'])
    
    direct_feat_sets = [
        ("No MF", ['Past_RV', 'return']),
        ("Full (With MF)", ['Past_RV', 'return'] + list(mfdfa_clean.columns))
    ]
    actual_10m_test = direct_merged['Future_Vol_24h'].values[4320:]
    test_dates_10m = direct_merged.index[4320:]

    for ai_name in ai_models:
        for feat_name, feats in direct_feat_sets:
            for is_tuned in [False, True]:
                tune_status = "Tuned" if is_tuned else "Untuned"
                eval_id = f"[Direct AI (10m)] | Model: {ai_name} ({tune_status}) | Feat: {feat_name}"
                
                if any(eval_id in line for line in completed_evals):
                    print(f"[Skip] {eval_id}")
                    continue
                
                params = tuned_10m.get(f"Direct_{ai_name}", None) if is_tuned else None
                if is_tuned and params is None: continue
                    
                preds = run_universal_ai_simulation(
                    direct_merged, feats, 'Future_Vol_24h', ai_name, params, train_window=4320, update_freq=INTRADAY_UPDATE_FREQ
                )
                rmse = np.sqrt(mean_squared_error(actual_10m_test, preds))
                log_result("Direct AI (10m)", ai_name, feat_name, is_tuned, rmse, "High Freq AI")
                save_predictions_to_csv(test_dates_10m, actual_10m_test, preds, "Direct AI (10m)", ai_name, feat_name, is_tuned)

    # ---------------------------------------------------------
    # [Phase 4] Hybrid AI (GARCH-MIDAS + 10min) 평가 
    # ---------------------------------------------------------
    print(f"\n▶ [Phase 4] Evaluating Hybrid AI (GARCH-MIDAS + AI - {current_year})")
    
    for g_name, exog_col in garch_scenarios.items():
        print(f"\n   >> Running Hybrid Scenario: {g_name}")
        try:
            df_base = monthly_garch_midas(df_10m, mfdfa_data, exog_col=exog_col)
            flat_data, features_all = prepare_flatten_features(df_base, mfdfa_data, lags=144)
            
            flat_data['GARCH_Daily'] = flat_data['Base_Vol'] * 12
            flat_data['Target_Residual'] = flat_data['Future_Vol_24h'] - flat_data['GARCH_Daily']
            
            hybrid_feat_sets = [
                ("No MF", ['Past_RV', 'return']),
                ("Full (With MF)", features_all)
            ]
            actual_hybrid_test = flat_data['Future_Vol_24h'].values[4320:]
            garch_daily_test = flat_data['GARCH_Daily'].values[4320:]
            test_dates_hybrid = flat_data.index[-len(actual_hybrid_test):]

            for ai_name in ai_models:
                for feat_name, feats in hybrid_feat_sets:
                    for is_tuned in [False, True]:
                        tune_status = "Tuned" if is_tuned else "Untuned"
                        scenario_name = f"Hybrid {g_name}"
                        eval_id = f"[{scenario_name}] | Model: {ai_name} ({tune_status}) | Feat: {feat_name}"
                        
                        if any(eval_id in line for line in completed_evals):
                            print(f"[Skip] {eval_id}")
                            continue
                        
                        save_key = f"Hybrid_{g_name}_{ai_name}"
                        params = tuned_10m.get(save_key, None) if is_tuned else None
                        
                        if is_tuned and params is None: continue
                            
                        pred_residuals = run_universal_ai_simulation(
                            flat_data, feats, 'Target_Residual', ai_name, params, train_window=4320, update_freq=INTRADAY_UPDATE_FREQ
                        )
                        
                        hybrid_preds = garch_daily_test + pred_residuals
                        
                        rmse = np.sqrt(mean_squared_error(actual_hybrid_test, hybrid_preds))
                        log_result(f"Hybrid {g_name}", ai_name, feat_name, is_tuned, rmse, "Correction AI")
                        save_predictions_to_csv(test_dates_hybrid, actual_hybrid_test, hybrid_preds, scenario_name, ai_name, feat_name, is_tuned)
                        
        except Exception as e:
            print(f"[Error] Hybrid {g_name} Failed: {e}")

    print("\nEvaluation Complete")


if __name__ == "__main__":
    years_to_run = [2020, 2021, 2022, 2023, 2024]
    freq_to_run = [1, 6, 144]
    
    for current_year in years_to_run:
        for current_freq in freq_to_run:
            TARGET_YEAR = current_year
            INTRADAY_UPDATE_FREQ = current_freq
            
            FINAL_RESULT_FILE = f"Final_RMSE_Results_{TARGET_YEAR}_Freq_{INTRADAY_UPDATE_FREQ}.txt"
            PREDICTIONS_CSV_FILE = f"Predictions_Log_{TARGET_YEAR}_Freq_{INTRADAY_UPDATE_FREQ}.csv"
            
            print("\n"+"="*30)
            print(f"Start Loop: {TARGET_YEAR}")
            print("="*30)
            try:
                target_df_10m = globals()[f"df_10m_{TARGET_YEAR}"]
                target_df_1d = globals()[f"df_1d_{TARGET_YEAR}"]
                target_mfdfa = globals()[f"mfdfa_{TARGET_YEAR}"]
                
                run_final_evaluation(target_df_10m, target_df_1d, target_mfdfa, TARGET_YEAR)
                
            except KeyError as e:
                print(f"\n[Error] No {TARGET_YEAR} Data in Memory")
                print(f"-> Missing Varialbes: {e}")
```

```python
TARGET_YEAR = "all"
TUNE_UPDATE_FREQ = 144
AI_N_TRIALS = 30

TUNING_SAVE_FILE = f"Tuning_Params_{TARGET_YEAR}_Freq_{TUNE_UPDATE_FREQ}.txt"

def tune_ai_hyperparameters(df_train, feature_cols, model_name, scenario_name, n_trials = 30, tune_freq = 144):

  # 고유 키 값을 저장하여 각 모델을 구분 (LSTM (No GARCH) VS GARCH-MIDAS + LSTM)
  save_key = f"{scenario_name}_{model_name}"

  # 저장된 결과 확인 및 스킵 로직
  saved_results = {}
  if os.path.exists(TUNING_SAVE_FILE):
    with open(TUNING_SAVE_FILE, 'r', encoding = 'utf-8') as f:
      try: saved_results = json.load(f)
      except: pass

  if save_key in saved_results:
    print(f"\n [{save_key}] alerady tuned. Skipping")
    return saved_results[save_key]

  print(f"\n[{save_key}] Optuna Optimization Start")

  df_tune = df_train.copy()
  
  if 'Base_Vol' in df_tune.columns:
    df_tune['GARCH_Daily'] = df_tune['Base_Vol'] * 12
    df_tune['Residual'] = df_tune['Future_Vol_24h'] - df_tune['GARCH_Daily']
    y_full = df_tune[['Residual']].values
  else:
    y_full = df_tune[['Future_Vol_24h']].values
    
  X_full = df_tune[feature_cols].values

  TRAIN_WINDOW = 4320
  start_idx = TRAIN_WINDOW
  end_idx = len(df_train)

  scaler_X = MinMaxScaler((-1, 1)).fit(X_full[:start_idx])
  scaler_y = MinMaxScaler((-1, 1)).fit(y_full[:start_idx])

  X_scaled = scaler_X.transform(X_full)
  y_scaled = scaler_y.transform(y_full)

  future_vol_arr = df_tune['Future_Vol_24h'].values
  if 'Base_Vol' in df_tune.columns:
    garch_daily_arr = df_tune['GARCH_Daily'].values
    
  def objective(trial):
    params = {}
    if model_name in ['LSTM', 'BI-LSTM', 'GRU', 'BI-GRU']:
      params['hidden_dim'] = trial.suggest_categorical('hidden_dim', [32, 64, 128])
      params['num_layers'] = trial.suggest_int('num_layers', 1, 2)
      params['learning_rate'] = trial.suggest_float('learning_rate', 1e-4, 5e-3, log=True)
      # params['dropout'] = trial.suggest_float('dropout', 0.0, 0.5)
    elif model_name == 'Ridge':
      params['alpha'] = trial.suggest_float('alpha', 0.1, 10.0)

    # 추후 업데이트 빈도를 수정할 때 해당 값을 수정해서 각 빈도에 맞는 파라미터 튜닝
    UPDATE_FREQ = tune_freq
    is_dl = model_name != 'Ridge'

    if is_dl:
      X_tensor = torch.FloatTensor(X_scaled).to(DEVICE).unsqueeze(1)
      y_tensor = torch.FloatTensor(y_scaled).to(DEVICE)
      input_dim = len(feature_cols)

      
      if model_name == 'LSTM':
        model = FlattenLSTM(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
      elif model_name == 'BI-LSTM':
        model = FlattenBiLSTM(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
      elif model_name == 'GRU':
        model = FlattenGRU(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
      elif model_name == 'BI-GRU':
        model = FlattenBiGRU(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)

      optimizer = torch.optim.Adam(model.parameters(), lr = params['learning_rate'])
      loss_fn = nn.HuberLoss(delta = 1.0).to(DEVICE)
    else:
      ridge_model = Ridge(alpha = params['alpha'])

    predictions, actuals = [], []

    pbar = tqdm(range(start_idx, end_idx), desc = f"{model_name} Tuning")

    for t in pbar:
      if (t - start_idx) % UPDATE_FREQ == 0:
        if is_dl:
          model.train()
          optimizer.zero_grad()
          pred = model(X_tensor[t - TRAIN_WINDOW : t])
          loss = loss_fn(pred, y_tensor[t - TRAIN_WINDOW : t])
          loss.backward()
          optimizer.step()
        else:
          ridge_model.fit(X_scaled[t - TRAIN_WINDOW : t], y_scaled[t - TRAIN_WINDOW : t])

      if is_dl:
        model.eval()
        with torch.no_grad():
          pred_val = model(X_tensor[t].unsqueeze(0)).item()
      else:
        pred_val = ridge_model.predict(X_scaled[t].reshape(1, -1)).item()

      predictions.append(pred_val)
      actuals.append(y_scaled[t][0])

      step_count = t - start_idx + 1
      if step_count > 0 and step_count % 4320 == 0:
        temp_pred_arr = np.array(predictions).reshape(-1, 1)
        temp_pred_rescaled = scaler_y.inverse_transform(temp_pred_arr).flatten()
        temp_actual_future = future_vol_arr[start_idx : start_idx + step_count]

        if 'Base_Vol' in df_tune.columns:
            # Hybrid: GARCH_Daily + 역변환된 예측 잔차
            temp_garch = garch_daily_arr[start_idx : start_idx + step_count]
            temp_final_pred = temp_garch + temp_pred_rescaled
        else:
            # Direct
            temp_final_pred = temp_pred_rescaled
        temp_rmse = np.sqrt(mean_squared_error(temp_actual_future, temp_final_pred))
        trial.report(temp_rmse, step_count // 4320)
        if trial.should_prune():
          raise optuna.TrialPruned()
    pred_array = np.array(predictions).reshape(-1, 1)
    pred_rescaled = scaler_y.inverse_transform(pred_array).flatten()
    actual_future = future_vol_arr[start_idx:end_idx]
    
    if 'Base_Vol' in df_tune.columns:
      final_pred_vol = garch_daily_arr[start_idx:end_idx] + pred_rescaled
    else:
      final_pred_vol = pred_rescaled
      
    rmse = np.sqrt(mean_squared_error(actual_future, final_pred_vol))

    return rmse

  pruner = optuna.pruners.MedianPruner(n_startup_trials = 5, n_warmup_steps = 2)
  study = optuna.create_study(direction = 'minimize', pruner = pruner)
  study.optimize(objective, n_trials = n_trials)

  saved_results[save_key] = study.best_params
  with open(TUNING_SAVE_FILE, 'w', encoding = 'utf-8') as f:
    json.dump(saved_results, f, indent = 4)

  print(f"[{save_key}] Best Performance: {study.best_value:.4f}")
  return study.best_params

if __name__ == "__main__":
    BEST_CONFIG = {"Direct": {}, "GARCH_K": {}, "Hybrid": {}}
    ai_model_to_tune = ["LSTM", "BI-LSTM", "GRU", "BI-GRU", "Ridge"]
    
    print("="*30)
    print(f"Start Parameter Tuning {TARGET_YEAR} (Freq: {TUNE_UPDATE_FREQ})")
    print("="*30)
    
    try:
        target_df_10m = globals()[f"df_10m_{TARGET_YEAR}"]
        target_mfdfa = globals()[f"mfdfa_{TARGET_YEAR}"]
        
        print(f"\n[Step 1] Tuning Direct AI Model")
        df_direct = target_df_10m.copy().sort_index()
        if df_direct.index.tz is not None: df_direct.index = df_direct.index.tz_localize(None)
        df_direct = df_direct[~df_direct.index.duplicated(keep='first')]

        df_direct['Future_Vol_24h'] = np.sqrt((df_direct['return']**2).rolling(window=144).sum().shift(-144))
        df_direct['Past_RV'] = np.sqrt((df_direct['return']**2).rolling(window=144).sum())
        df_direct = df_direct.dropna()

        mfdfa_clean = target_mfdfa.copy()
        if mfdfa_clean.index.tz is not None: mfdfa_clean.index = mfdfa_clean.index.tz_localize(None)
        mfdfa_clean = mfdfa_clean[~mfdfa_clean.index.duplicated(keep='first')]

        df_direct['temp_date'] = df_direct.index.normalize()
        df_merged = df_direct.merge(mfdfa_clean, left_on='temp_date', right_index=True, how='left', validate='m:1')
        df_merged = df_merged.fillna(method='ffill').dropna().drop(columns=['temp_date'])

        feat_cols = ['Past_RV', 'return'] + list(mfdfa_clean.columns)

        for ai_name in ai_model_to_tune:
            best_p = tune_ai_hyperparameters(df_merged, feat_cols, ai_name, "Direct", n_trials=AI_N_TRIALS, tune_freq=TUNE_UPDATE_FREQ)
            BEST_CONFIG["Direct"][ai_name] = best_p

        garch_scenarios = {"Benchmark": None, "Delta_H": "delta_H", "Delta_Alpha": "delta_alpha"}

        for sc_name, exog_col in garch_scenarios.items():
            print(f"\n▶ [STEP 2] Processing Scenario: {sc_name}")

            best_k = tune_garch_midas_k(target_df_10m, target_mfdfa, exog_col, n_trials=15)
            BEST_CONFIG["GARCH_K"][sc_name] = best_k

            df_base = monthly_garch_midas(target_df_10m, target_mfdfa, exog_col=exog_col, midas_k=best_k)
            flat_data, features = prepare_flatten_features(df_base, target_mfdfa, lags=144)

            print(f"\n▶ [STEP 3] Tuning Hybrid AI for {sc_name}")
            BEST_CONFIG["Hybrid"][sc_name] = {}

            for ai_name in ai_model_to_tune:
                best_p = tune_ai_hyperparameters(flat_data, features, ai_name, f"Hybrid_{sc_name}", n_trials=AI_N_TRIALS, tune_freq=TUNE_UPDATE_FREQ)
                BEST_CONFIG["Hybrid"][sc_name][ai_name] = best_p

        print("\n" + "="*80)
        print(f"🎉 FINAL OPTIMIZED CONFIGURATION FOR {TARGET_YEAR} (Freq: {TUNE_UPDATE_FREQ})")
        print("="*80)
        print(json.dumps(BEST_CONFIG, indent=4))

    except Exception as e:
        print(f"🚨 Tuning Process Failed: {e}")
```

```python
# 10분봉 데이터 업데이트 빈도 (1 = 10분 마다, 6 = 1시간 마다, 144 = 1일 마다)
INTRADAY_UPDATE_FREQ = 1

# 일별 데이터 업데이트 주기
DAILY_UPDATE_FREQ = 1

FINAL_RESULT_FILE = ""
PREDICTIONS_CSV_FILE = ""

TUNING_10M_FILE = "Tuning_Params_2020_Freq_144.txt"
TUNING_DAILY_FILE = "Daily_Direct_Tuning_Params_2020.txt"

def log_result(scenario, model, feature_set, is_tuned, rmse, note):
    tune_status = "Tuned" if is_tuned else "Untuned"
    result_str = f"[{scenario}] | Model: {model} ({tune_status}) | Feat: {feature_set} | RMSE: {rmse:.4f} | Note: {note}\n"
    print(f"{result_str.strip()}")
    
    with open(FINAL_RESULT_FILE, 'a', encoding='utf-8') as f:
        f.write(result_str)

def save_predictions_to_csv(dates, actuals, preds, scenario, model_name, feature_name, is_tuned):
    tune_status = "Tuned" if is_tuned else "Untuned"
    full_model_name = f"[{scenario}] {model_name} ({tune_status})"
    
    df_log = pd.DataFrame({
        'Date': dates,
        'Actual_RV': actuals,
        'Predicted_RV': preds,
        'Model_Name': full_model_name,
        'Feature': feature_name
    })
    
    file_exists = os.path.isfile(PREDICTIONS_CSV_FILE)
    df_log.to_csv(PREDICTIONS_CSV_FILE, mode = 'a', header = not file_exists, index = False)
   
def run_universal_ai_simulation(df, feature_cols, target_col, model_name, params, train_window, update_freq):
    X_full = df[feature_cols].values
    y_full = df[[target_col]].values
    
    start_idx = train_window
    end_idx = len(df)
    
    scaler_X = MinMaxScaler((-1, 1)).fit(X_full[:start_idx])
    scaler_y = MinMaxScaler((-1, 1)).fit(y_full[:start_idx])
    
    X_scaled = scaler_X.transform(X_full)
    y_scaled = scaler_y.transform(y_full)
    
    if params is None:
        if model_name != 'Ridge':
            params = {'hidden_dim': 64, 'num_layers': 2, 'learning_rate': 0.001}
        else:
            params = {'alpha': 1.0}
            
    is_dl = model_name != 'Ridge'
    
    if is_dl:
        X_tensor = torch.FloatTensor(X_scaled).to(DEVICE).unsqueeze(1)
        y_tensor = torch.FloatTensor(y_scaled).to(DEVICE)
        input_dim = len(feature_cols)
        
        if model_name == 'LSTM': model = FlattenLSTM(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
        elif model_name == 'BI-LSTM': model = FlattenBiLSTM(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
        elif model_name == 'GRU': model = FlattenGRU(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
        elif model_name == 'BI-GRU': model = FlattenBiGRU(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
            
        optimizer = torch.optim.Adam(model.parameters(), lr=params['learning_rate'])
        loss_fn = nn.HuberLoss(delta = 1.0).to(DEVICE)
    else:
        ridge_model = Ridge(alpha=params['alpha'])

    predictions = []
    
    pbar = tqdm(range(start_idx, end_idx), desc=f"Evaluating {model_name}", leave=False)
    for t in pbar:
        if (t - start_idx) % update_freq == 0:
            if is_dl:
                model.train()
                optimizer.zero_grad()
                pred = model(X_tensor[t - train_window : t])
                loss = loss_fn(pred, y_tensor[t - train_window : t])
                loss.backward()
                optimizer.step()
            else:
                ridge_model.fit(X_scaled[t - train_window : t], y_scaled[t - train_window : t])
                
        if is_dl:
            model.eval()
            with torch.no_grad():
                pred_val = model(X_tensor[t].unsqueeze(0)).item()
        else:
            pred_val = ridge_model.predict(X_scaled[t].reshape(1, -1)).item()
            
        predictions.append(pred_val)
        
    pred_array = np.array(predictions).reshape(-1, 1)
    pred_rescaled = scaler_y.inverse_transform(pred_array).flatten()
    
    return pred_rescaled

def run_final_evaluation(df_10m, df_daily, mfdfa_data, current_year):
    print("\n" + "="*70)
    print(f"FINAL EVALUATION SUITE (Year: {current_year} / Update Freq: {INTRADAY_UPDATE_FREQ})")
    print(f"Saving to: {FINAL_RESULT_FILE}")
    print("="*70)
    
    if not os.path.exists(FINAL_RESULT_FILE):
        with open(FINAL_RESULT_FILE, 'w', encoding='utf-8') as f:
            f.write(f"--- Evaluation for Year: {current_year} | Freq: {INTRADAY_UPDATE_FREQ} ---\n")
            f.write("Scenario | Model | Feature Set | RMSE | Note\n")
            f.write("-" * 80 + "\n")
    
    completed_evals = set()
    with open(FINAL_RESULT_FILE, 'r', encoding = 'utf-8') as f:
        for line in f:
            if "|" in line and "RMSE" in line:
                completed_evals.add(line.strip())

    garch_scenarios = {
        "Benchmark": None,
        "Delta_H": "delta_H",
        "Delta_Alpha": "delta_alpha"
    }
    
    ai_models = ["LSTM", "BI-LSTM", "GRU", "BI-GRU", "Ridge"]
    
    tuned_10m = {}
    tuned_daily = {}
    try:
        with open(TUNING_10M_FILE, 'r') as f: tuned_10m = json.load(f)
        with open(TUNING_DAILY_FILE, 'r') as f: tuned_daily = json.load(f)
    except Exception as e:
        print(f"[Warning] Tuning file load failed: {e}. Tuned runs will be skipped.")

    print(f"\n[Phase 1] Evaluating Baseline GARCH Models ({current_year})")
    
    df_temp = monthly_garch_midas(df_10m, mfdfa_data, exog_col=None)
    flat_temp, _ = prepare_flatten_features(df_temp, mfdfa_data, lags=144)
    TEST_DAYS_10M = (len(flat_temp) - 4320) // 144
    actual_rv_series = flat_temp['Future_Vol_24h']
    
    # 1. Daily & Intraday 기본 GARCH
    if not any("[1. Baseline]" in line for line in completed_evals):
        rmse_daily = run_daily_garch(df_daily, actual_rv_series, TEST_DAYS_10M)
        log_result("1. Baseline", "GARCH(1,1)", "Daily Return", False, rmse_daily, "Low Freq")
    else:
        print("[Skip] [1. Baseline] GARCH(1, 1)")
        
    if not any("[2. Baseline]" in line for line in completed_evals):
        rmse_intra = run_intraday_garch(df_10m, actual_rv_series, TEST_DAYS_10M)
        log_result("2. Baseline", "GARCH(1,1)", "10m Return", False, rmse_intra, "High Freq")
    else:
        print("[Skip] [2. Baseline] GARCH(1, 1)")
    
    # 2. GARCH-MIDAS 시나리오별 평가 (Benchmark, Delta_H, Delta_Alpha)
    for g_name, exog_col in garch_scenarios.items():
        eval_id = f"[3. Baseline GM] | Model: GARCH-MIDAS ({g_name})"
        if any(eval_id in line for line in completed_evals):
            print(f"[Skip] {eval_id}")
            continue
        
        try:
            df_base = monthly_garch_midas(df_10m, mfdfa_data, exog_col=exog_col)
            flat_data, _ = prepare_flatten_features(df_base, mfdfa_data, lags=144)
            garch_midas_pred = flat_data['Base_Vol'][-TEST_DAYS_10M*144:] * 12
            actual_rv_test = flat_data['Future_Vol_24h'][-TEST_DAYS_10M*144:]
            
            rmse_gm = np.sqrt(mean_squared_error(actual_rv_test, garch_midas_pred))
            log_result(f"3. Baseline GM", f"GARCH-MIDAS ({g_name})", f"10m + {exog_col}", False, rmse_gm, "Macro")
        except Exception as e:
            print(f"[Error] Baseline GARCH-MIDAS ({g_name}) Failed: {e}")


    # ---------------------------------------------------------
    # [Phase 2] Direct AI (Daily) 평가
    # ---------------------------------------------------------
    print(f"\n▶ [Phase 2] Evaluating Direct AI (Daily Data - {current_year})")
    daily_data = prepare_daily_direct_data(df_10m, df_daily)
    daily_feats = ['Past_RV', 'daily_return']
    target_daily = 'Future_Vol_1d'
    actual_daily_test = daily_data[target_daily].values[30:]
    test_dates_daily = daily_data.index[30:]
    
    for ai_name in ai_models:
        for is_tuned in [False, True]:
            tune_status = "Tuned" if is_tuned else "Untuned"
            eval_id = f"[Direct AI (Daily)] | Model: {ai_name} ({tune_status}) | Feat: Daily Feat"
            
            if any(eval_id in line for line in completed_evals):
                print(f"[Skip] {eval_id} (Already calculated)")
                continue
            
            params = tuned_daily.get(f"Daily_Direct_{ai_name}", None) if is_tuned else None
            if is_tuned and params is None: continue 
                
            preds = run_universal_ai_simulation(
                daily_data, daily_feats, target_daily, ai_name, params, train_window=30, update_freq=DAILY_UPDATE_FREQ
            )
            rmse = np.sqrt(mean_squared_error(actual_daily_test, preds))
            log_result("Direct AI (Daily)", ai_name, "Daily Feat", is_tuned, rmse, "Low Freq AI")
            save_predictions_to_csv(test_dates_daily, actual_daily_test, preds, "Direct AI (Daily)", ai_name, "Daily Feat", is_tuned)

    # ---------------------------------------------------------
    # [Phase 3] Direct AI (10min) 평가
    # ---------------------------------------------------------
    print(f"\n▶ [Phase 3] Evaluating Direct AI (10min Data - {current_year})")
    df_direct = df_10m.copy()
    df_direct.index = df_direct.index.tz_localize(None)
    df_direct = df_direct[~df_direct.index.duplicated(keep='first')]
    df_direct['Future_Vol_24h'] = (df_direct['return']**2).rolling(window=144).sum().shift(-144).apply(np.sqrt)
    df_direct['Past_RV'] = (df_direct['return']**2).rolling(window=144).sum().apply(np.sqrt)
    df_direct = df_direct.dropna()
    
    mfdfa_clean = mfdfa_data.copy()
    mfdfa_clean.index = mfdfa_clean.index.tz_localize(None)
    mfdfa_clean = mfdfa_clean[~mfdfa_clean.index.duplicated(keep='first')]
    
    df_direct['temp_date'] = df_direct.index.normalize()
    direct_merged = df_direct.merge(mfdfa_clean, left_on='temp_date', right_index=True, how='left')
    direct_merged = direct_merged.fillna(method='ffill').dropna().drop(columns=['temp_date'])
    
    direct_feat_sets = [
        ("No MF", ['Past_RV', 'return']),
        ("Full (With MF)", ['Past_RV', 'return'] + list(mfdfa_clean.columns))
    ]
    actual_10m_test = direct_merged['Future_Vol_24h'].values[4320:]
    test_dates_10m = direct_merged.index[4320:]

    for ai_name in ai_models:
        for feat_name, feats in direct_feat_sets:
            for is_tuned in [False, True]:
                tune_status = "Tuned" if is_tuned else "Untuned"
                eval_id = f"[Direct AI (10m)] | Model: {ai_name} ({tune_status}) | Feat: {feat_name}"
                
                if any(eval_id in line for line in completed_evals):
                    print(f"[Skip] {eval_id}")
                    continue
                
                params = tuned_10m.get(f"Direct_{ai_name}", None) if is_tuned else None
                if is_tuned and params is None: continue
                    
                preds = run_universal_ai_simulation(
                    direct_merged, feats, 'Future_Vol_24h', ai_name, params, train_window=4320, update_freq=INTRADAY_UPDATE_FREQ
                )
                rmse = np.sqrt(mean_squared_error(actual_10m_test, preds))
                log_result("Direct AI (10m)", ai_name, feat_name, is_tuned, rmse, "High Freq AI")
                save_predictions_to_csv(test_dates_10m, actual_10m_test, preds, "Direct AI (10m)", ai_name, feat_name, is_tuned)

    # ---------------------------------------------------------
    # [Phase 4] Hybrid AI (GARCH-MIDAS + 10min) 평가 
    # ---------------------------------------------------------
    print(f"\n▶ [Phase 4] Evaluating Hybrid AI (GARCH-MIDAS + AI - {current_year})")
    
    for g_name, exog_col in garch_scenarios.items():
        print(f"\n   >> Running Hybrid Scenario: {g_name}")
        try:
            df_base = monthly_garch_midas(df_10m, mfdfa_data, exog_col=exog_col)
            flat_data, features_all = prepare_flatten_features(df_base, mfdfa_data, lags=144)
            
            flat_data['GARCH_Daily'] = flat_data['Base_Vol'] * 12
            flat_data['Target_Residual'] = flat_data['Future_Vol_24h'] - flat_data['GARCH_Daily']
            
            hybrid_feat_sets = [
                ("No MF", ['Past_RV', 'return']),
                ("Full (With MF)", features_all)
            ]
            actual_hybrid_test = flat_data['Future_Vol_24h'].values[4320:]
            garch_daily_test = flat_data['GARCH_Daily'].values[4320:]
            test_dates_hybrid = flat_data.index[-len(actual_hybrid_test):]

            for ai_name in ai_models:
                for feat_name, feats in hybrid_feat_sets:
                    for is_tuned in [False, True]:
                        tune_status = "Tuned" if is_tuned else "Untuned"
                        scenario_name = f"Hybrid {g_name}"
                        eval_id = f"[{scenario_name}] | Model: {ai_name} ({tune_status}) | Feat: {feat_name}"
                        
                        if any(eval_id in line for line in completed_evals):
                            print(f"[Skip] {eval_id}")
                            continue
                        
                        save_key = f"Hybrid_{g_name}_{ai_name}"
                        params = tuned_10m.get(save_key, None) if is_tuned else None
                        
                        if is_tuned and params is None: continue
                            
                        pred_residuals = run_universal_ai_simulation(
                            flat_data, feats, 'Target_Residual', ai_name, params, train_window=4320, update_freq=INTRADAY_UPDATE_FREQ
                        )
                        
                        hybrid_preds = garch_daily_test + pred_residuals
                        
                        rmse = np.sqrt(mean_squared_error(actual_hybrid_test, hybrid_preds))
                        log_result(f"Hybrid {g_name}", ai_name, feat_name, is_tuned, rmse, "Correction AI")
                        save_predictions_to_csv(test_dates_hybrid, actual_hybrid_test, hybrid_preds, scenario_name, ai_name, feat_name, is_tuned)
                        
        except Exception as e:
            print(f"[Error] Hybrid {g_name} Failed: {e}")

    print("\nEvaluation Complete")


if __name__ == "__main__":
    years_to_run = [2020]
    freq_to_run = [1]
    
    for current_year in years_to_run:
        for current_freq in freq_to_run:
            TARGET_YEAR = current_year
            INTRADAY_UPDATE_FREQ = current_freq
            
            FINAL_RESULT_FILE = f"Final_RMSE_Results_{TARGET_YEAR}_Freq_{INTRADAY_UPDATE_FREQ}.txt"
            PREDICTIONS_CSV_FILE = f"Predictions_Log_{TARGET_YEAR}_Freq_{INTRADAY_UPDATE_FREQ}.csv"
            
            print("\n"+"="*30)
            print(f"Start Loop: {TARGET_YEAR}")
            print("="*30)
            try:
                target_df_10m = globals()[f"df_10m_{TARGET_YEAR}"]
                target_df_1d = globals()[f"df_1d_{TARGET_YEAR}"]
                target_mfdfa = globals()[f"mfdfa_{TARGET_YEAR}"]
                
                run_final_evaluation(target_df_10m, target_df_1d, target_mfdfa, TARGET_YEAR)
                
            except KeyError as e:
                print(f"\n[Error] No {TARGET_YEAR} Data in Memory")
                print(f"-> Missing Varialbes: {e}")
```

```python
MONTHLY_TUNING_FILE = "Monthly_Tuning_Params_2021.json"

def tune_garch_midas_single_month(df_train_10m, mfdfa_data, exog_col, n_trials = 7):
    print(f"Tuning GARCH-MIDAS (exog: {exog_col})")
    
    split_idx = int(len(df_train_10m) * 0.8)
    train_slice = df_train_10m.iloc[:split_idx].copy()
    test_slice = df_train_10m.iloc[split_idx:].copy()
    
    daily_rv_all = (df_train_10m['return']**2).resample('1D').sum().to_frame('rv')
    
    if exog_col is None:
        df_x_full = pd.DataFrame(0, index = daily_rv_all.index, columns = ['exog_x'])
    else:
        df_x_full = mfdfa_data[[exog_col]].copy()
        df_x_full.columns = ['exog_x']
        df_x_full = df_x_full.resample('1D').mean() # -> 여기 이해 안됨
        df_x_full = df_x_full.reindex(daily_rv_all.index).fillna(method = 'ffill').fillna(0)
        
    test_slice['Future_RV_24h'] = (test_slice['return']**2).rolling(window = 144).sum().shift(-144)
    test_slice['Future_Vol_24h'] = np.sqrt(test_slice['Future_RV_24h'])
        
    def objective(trial):
        k = trial.suggest_int('midas_k', 1, 7)
        try:
            pred_df = garch_midas(train_slice, test_slice, daily_rv_all, df_x_full, midas_k = k)
            
            val_df = test_slice.copy()
            val_df['gm_vol'] = pred_df['gm_vol']
            val_df_clean = val_df.dropna(subset=['Future_Vol_24h', 'gm_vol'])
            
            actual_vol = val_df_clean['Future_Vol_24h']
            pred_vol = val_df_clean['gm_vol'] * 12
            
            rmse = np.sqrt(mean_squared_error(actual_vol, pred_vol))
            return rmse
        except Exception:
            raise optuna.exceptions.TrialPruned()
        
    study = optuna.create_study(direction = 'minimize', sampler = TPESampler(seed = 42))
    study.optimize(objective, n_trials = n_trials, show_progress_bar = False)
    
    try: best_k = study.best_params['midas_k']
    except: best_k = 3
    
    print(f"Best K: {best_k} (RMSE: {study.best_value:.4f})")
    return best_k

def tune_ai_single_month(df_train, feature_cols, model_name, scenario_type, n_trials = 50):
    print(f"Tuning {model_name} | Scenario {scenario_type}")
    
    is_daily = 'Daily' in scenario_type
    is_hybrid = 'Hybrid' in scenario_type
    
    ROLL_WINDOW = 7 if is_daily else 144
    UPDATE_FREQ = 1
    DELAY = 1 if is_daily else 144
    
    split_idx = int(len(df_train) * 0.8)
    
    target_col = 'Future_Vol_1d' if is_daily else 'Future_Vol_24h'
    if is_hybrid:
        if 'Target_Residual' not in df_train.columns:
            df_train['GARCH_Daily'] = df_train['Base_Vol'] * 12
            df_train['Target_Residual'] = df_train['Future_Vol_24h'] - df_train['GARCH_Daily'] # -> GARCH_Daily가 GARCH-MIDAS로 예측한 RV가 맞는지 확인
        y_full = df_train[['Target_Residual']].values
    else:
        y_full = df_train[[target_col]].values
        
    X_full = df_train[feature_cols].values
    
    scaler_X = MinMaxScaler((-1, 1)).fit(X_full[:split_idx])
    scaler_y = MinMaxScaler((-1, 1)).fit(y_full[:split_idx])
    
    X_scaled = scaler_X.transform(X_full)
    y_scaled = scaler_y.transform(y_full)
    
    actual_val_future = df_train[target_col].values[split_idx:]
    if is_hybrid:
        base_vol_val = (df_train['Base_Vol'] * 12).values[split_idx:]
             
    def objective(trial):
        params = {}
        
        is_dl = model_name != 'Ridge'
        
        if is_dl:
            params['hidden_dim'] = trial.suggest_categorical('hidden_dim', [32, 64, 128])
            params['num_layers'] = trial.suggest_int('num_layers', 1, 2)
            params['learning_rate'] = trial.suggest_float('learning_rate', 1e-4, 5e-3, log = True)
        else:
            params['alpha'] = trial.suggest_float('alpha', 0.1, 10.0)
        
        if is_dl:
            lags = 144
            
            if len(feature_cols) >= lags:
                num_features = len(feature_cols) // lags
                X_3d = X_scaled.reshape(-1, lags, num_features)
                X_3d = np.flip(X_3d, axis=1).copy()
            else:
                num_features = len(feature_cols)
                X_3d = np.zeros((len(X_scaled), lags, num_features))
                for i in range(lags - 1, len(X_scaled)):
                    X_3d[i] = X_scaled[i - lags + 1 : i + 1]
            
            X_tensor = torch.FloatTensor(X_3d).to(DEVICE)
            y_tensor = torch.FloatTensor(y_scaled).to(DEVICE)
            input_dim = num_features
            
            if model_name == 'LSTM': model = FlattenLSTM(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
            elif model_name == 'BI-LSTM': model = FlattenBiLSTM(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
            elif model_name == 'GRU': model = FlattenGRU(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
            elif model_name == 'BI-GRU': model = FlattenBiGRU(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
            
            optimizer = torch.optim.Adam(model.parameters(), lr = params['learning_rate'])
            loss_fn = nn.HuberLoss(delta = 1.0).to(DEVICE)
        else:
            ridge_model = Ridge(alpha = params['alpha'])
            
        val_predictions = []
        
        for t in range(ROLL_WINDOW, len(df_train)):
            if (t - ROLL_WINDOW) % UPDATE_FREQ == 0:
                if t >= ROLL_WINDOW + DELAY:
                    train_end = t - DELAY
                    train_start = train_end - ROLL_WINDOW
                    if is_dl:
                        model.train()
                        optimizer.zero_grad()
                        pred = model(X_tensor[t - ROLL_WINDOW : t])
                        loss = loss_fn(pred, y_tensor[t - ROLL_WINDOW : t])
                        loss.backward()
                        optimizer.step()
                    else:
                        ridge_model.fit(X_scaled[t - ROLL_WINDOW : t], y_scaled[t - ROLL_WINDOW : t])
                        
            if t >= split_idx:
                if is_dl:
                    model.eval()
                    with torch.no_grad():
                        pred_val = model(X_tensor[t].unsqueeze(0)).item()
                else:
                    pred_val = ridge_model.predict(X_scaled[t].reshape(1, -1)).item()
                val_predictions.append(pred_val)
                
        val_preds_arr = np.array(val_predictions).reshape(-1, 1)
        val_preds = scaler_y.inverse_transform(val_preds_arr).flatten()
        
        if is_hybrid: final_pred_vol = val_preds + base_vol_val
        else: final_pred_vol = val_preds
        
        rmse = np.sqrt(mean_squared_error(actual_val_future, final_pred_vol))
        return rmse
    
    study = optuna.create_study(direction = 'minimize', sampler = TPESampler(seed = 42))
    study.optimize(objective, n_trials = n_trials, show_progress_bar = False)
    
    return study.best_params
```

```python
def run_oos_simulation_for_month(df_period, test_start, feature_cols, model_name, params, scenario_type):
    is_daily = 'Daily' in scenario_type
    is_hybrid = 'Hybrid' in scenario_type
    
    ROLL_WINDOW = 7 if is_daily else 144
    UPDATE_FREQ = 1
    DELAY = 1 if is_daily else 144
    
    train_len = len(df_period[df_period.index < test_start])
    
    target_col = 'Future_Vol_1d' if is_daily else 'Future_Vol_24h'
    
    if is_hybrid:
        if 'GARCH_Daily' not in df_period.columns:
            df_period['GARCH_Daily'] = df_period['Base_Vol'] * 12
        if 'Target_Residual' not in df_period.columns:
            df_period['Target_Residual'] = df_period['Future_Vol_24h'] - df_period['GARCH_Daily']
        y_full = df_period[['Target_Residual']].values
        base_vol_test = (df_period['Base_Vol'] * 12).values[train_len:]
    else:
        y_full = df_period[[target_col]].values
        
    X_full = df_period[feature_cols].values
    
    scaler_X = MinMaxScaler((-1, 1)).fit(X_full[:train_len])
    scaler_y = MinMaxScaler((-1, 1)).fit(y_full[:train_len])
    
    X_scaled = scaler_X.transform(X_full)
    y_scaled = scaler_y.transform(y_full)
    
    is_dl = model_name != 'Ridge'
    
    if is_dl:
        lags = 144
        
        if len(feature_cols) >= lags:
            num_features = len(feature_cols) // lags
            X_3d = X_scaled.reshape(-1, lags, num_features)
            X_3d = np.flip(X_3d, axis=1).copy()
        else:
            num_features = len(feature_cols)
            X_3d = np.zeros((len(X_scaled), lags, num_features))
            for i in range(lags - 1, len(X_scaled)):
                X_3d[i] = X_scaled[i - lags + 1 : i + 1]
        
        X_tensor = torch.FloatTensor(X_3d).to(DEVICE)
        y_tensor = torch.FloatTensor(y_scaled).to(DEVICE)
        input_dim = num_features 
        
        if model_name == 'LSTM': model = FlattenLSTM(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
        elif model_name == 'BI-LSTM': model = FlattenBiLSTM(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
        elif model_name == 'GRU': model = FlattenGRU(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)
        elif model_name == 'BI-GRU': model = FlattenBiGRU(input_dim, params['hidden_dim'], params['num_layers']).to(DEVICE)

        optimizer = torch.optim.Adam(model.parameters(), lr=params['learning_rate'])
        loss_fn = nn.HuberLoss(delta = 1.0).to(DEVICE)
    else:
        ridge_model = Ridge(alpha=params['alpha'])

    predictions = []
    actuals = df_period[target_col].values[train_len:]
    test_dates = df_period.index[train_len:]

    for t in range(ROLL_WINDOW, len(df_period)):
        is_test_period = t >= train_len

        if is_test_period:
            if is_dl:
                model.eval()
                with torch.no_grad():
                    pred_val = model(X_tensor[t].unsqueeze(0)).item()
            else:
                pred_val = ridge_model.predict(X_scaled[t].reshape(1, -1)).item()
            predictions.append(pred_val)

        if (t - ROLL_WINDOW) % UPDATE_FREQ == 0:
            if t >= ROLL_WINDOW + DELAY:
                train_end = t - DELAY
                train_start = train_end - ROLL_WINDOW
                
                if is_dl:
                    model.train()
                    optimizer.zero_grad()
                    pred_train = model(X_tensor[t - ROLL_WINDOW : t])
                    loss = loss_fn(pred_train, y_tensor[t - ROLL_WINDOW : t])
                    loss.backward()
                    optimizer.step()
                else:
                    ridge_model.fit(X_scaled[t - ROLL_WINDOW : t], y_scaled[t - ROLL_WINDOW : t])

    val_preds_arr = np.array(predictions).reshape(-1, 1)
    val_preds = scaler_y.inverse_transform(val_preds_arr).flatten()

    if is_hybrid: final_pred_vol = val_preds + base_vol_test
    else:         final_pred_vol = val_preds

    return test_dates, actuals, final_pred_vol


def run_annual_oos_evaluation(monthly_experiments, df_full, feature_cols, model_name, scenario_type, monthly_params_dict):
    print(f"\n{'='*60}")
    print(f"Starting Annual OOS Evaluation: {model_name} ({scenario_type})")
    print(f"{'='*60}")
    
    all_dates = []
    all_actuals = []
    all_preds = []
    
    for exp in monthly_experiments:
        target_month = exp['target_month']  
        test_start = exp['test_start']      
        
        print(f"Processing {target_month} ...", end=" ")
        
        if target_month in monthly_params_dict:
            best_p = monthly_params_dict[target_month]
        else:
            print("[Warning] No tuned params found. Using defaults.", end=" ")
            best_p = {'hidden_dim': 64, 'num_layers': 1, 'learning_rate': 0.001} if model_name != 'Ridge' else {'alpha': 1.0}
            
        df_period = df_full.loc[exp['train_start'] : exp['test_end']].copy()
        
        dates, actuals, preds = run_oos_simulation_for_month(
            df_period=df_period,
            test_start=test_start,
            feature_cols=feature_cols,
            model_name=model_name,
            params=best_p,
            scenario_type=scenario_type
        )
        
        all_dates.extend(dates)
        all_actuals.extend(actuals)
        all_preds.extend(preds)
        print("Done!")
        
    final_rmse = np.sqrt(mean_squared_error(all_actuals, all_preds))
    print(f"\nFinal OOS RMSE for {model_name} ({scenario_type}): {final_rmse:.4f}")
    
    df_results = pd.DataFrame({
        'Datetime': all_dates,
        'Actual_RV': all_actuals,
        'Predicted_RV': all_preds,
        'Model_Name': model_name,
        'Scenario': scenario_type
    })
    
    return final_rmse, df_results
```

### 실험 1. Scaler 변경
- 기존에 사용하던 MinMaxScaler를 사용시 GARCH-MIDAS + AI 모델들의 Min과 Max 값이 실제 값과 엄청난 차이가 발생
- MinMaxScaler를 사용하여 scaling을 진행했기 때문에 Standard Scaler로 실험 <br>
**파라미터는 다음과 같이 고정**
- alpha = 1.0
- hidden_dim = 64
- num_layers = 2
- learning_rate = 0.002
- midas_k = 7

### 실험 2. Loss Function 변경
- 기존에 사용하던 HuberLoss가 아닌 Asymmetric Loss를 사용
- loss 함수가 계산하는 error = (실제 RV - GARCH 예측 RV) - AI 모델의 잔차 예측
  - error = y_true - y_pred

```python
BASELINE_PARAMS = {
    "Ridge": {"alpha": 1.0},
    "LSTM": {"hidden_dim": 64, "num_layers": 2, "learning_rate": 0.002},
    "BI-LSTM": {"hidden_dim": 64, "num_layers": 2, "learning_rate": 0.002},
    "GRU": {"hidden_dim": 64, "num_layers": 2, "learning_rate": 0.002},
    "BI-GRU": {"hidden_dim": 64, "num_layers": 2, "learning_rate": 0.002},
    "GARCH_K": 30
}

def run_oos_simulation_for_month(df_period, test_start, feature_cols, model_name, params, scenario_type, dropout_rate = 0.0):
    is_daily = 'Daily' in scenario_type
    is_hybrid = 'Hybrid' in scenario_type
    
    train_len = len(df_period[df_period.index < test_start])
    
    ROLL_WINDOW = 7 if is_daily else train_len
    UPDATE_FREQ = 144
    DELAY = 1 if is_daily else 144
    
    
    target_col = 'Target_Residual' if is_hybrid else 'Future_Vol_24h'
    y_full = df_period[[target_col]].values
    
    if is_hybrid:
        base_vol_test = (df_period['Base_Vol_24h']).values[train_len:]
    
    X_full = df_period[feature_cols].values
        
    safe_train_end = train_len - DELAY
    scaler_X = StandardScaler().fit(X_full[:safe_train_end])
    scaler_y = StandardScaler().fit(y_full[:safe_train_end])
    
    X_scaled = scaler_X.transform(X_full)
    y_scaled = scaler_y.transform(y_full)
    
    is_dl = model_name != 'Ridge'
    lags = 144
    
    if is_dl:
        num_features = len(feature_cols)
        
        X_3d = np.zeros((len(X_scaled), lags, num_features))
        for i in range(lags - 1, len(X_scaled)):
            X_3d[i] = X_scaled[i - lags + 1 : i + 1]
                
        X_tensor = torch.FloatTensor(X_3d).to(DEVICE)
        y_tensor = torch.FloatTensor(y_scaled).to(DEVICE)
        input_dim = num_features
        
        if model_name == 'LSTM': model = FlattenLSTM(input_dim, params['hidden_dim'], params['num_layers'], dropout=dropout_rate).to(DEVICE)
        elif model_name == 'BI-LSTM': model = FlattenBiLSTM(input_dim, params['hidden_dim'], params['num_layers'], dropout=dropout_rate).to(DEVICE)
        elif model_name == 'GRU': model = FlattenGRU(input_dim, params['hidden_dim'], params['num_layers'], dropout=dropout_rate).to(DEVICE)
        elif model_name == 'BI-GRU': model = FlattenBiGRU(input_dim, params['hidden_dim'], params['num_layers'], dropout=dropout_rate).to(DEVICE)

        optimizer = torch.optim.Adam(model.parameters(), lr=params['learning_rate'])
        loss_fn = nn.HuberLoss(delta = 1.0).to(DEVICE)
        
        print(f"[{model_name}] Initial Pre-Training (10 epochs)")
        model.train()
        pretrain_end = train_len -DELAY
        if pretrain_end > lags:
            batch_size = 256  
            X_pre = X_tensor[lags:pretrain_end]
            y_pre = y_tensor[lags:pretrain_end]
            dataset_size = len(X_pre)
            for epoch in range(10):
                for i in range(0, dataset_size, batch_size):
                    X_batch = X_pre[i : i + batch_size]
                    y_batch = y_pre[i : i + batch_size]
                    optimizer.zero_grad()
                    pred_train = model(X_batch)
                    loss = loss_fn(pred_train, y_batch)
                    loss.backward()
                    optimizer.step()
    else:
        ridge_model = Ridge(alpha=params['alpha'])
        pretrain_end = train_len - DELAY
        if pretrain_end > lags:
            ridge_model.fit(X_scaled[lags:pretrain_end], y_scaled[lags:pretrain_end])

    predictions = []
    actuals = df_period['Future_Vol_24h'].values[train_len:]
    test_dates = df_period.index[train_len:]

    for t in tqdm(range(train_len, len(df_period)), desc=f"{model_name} Rolling", leave=False):
        
        if is_dl:
            model.eval()
            with torch.no_grad():
                pred_val = model(X_tensor[t].unsqueeze(0)).item()
        else:
            pred_val = ridge_model.predict(X_scaled[t].reshape(1, -1)).item()
        predictions.append(pred_val)

        if (t - train_len) % UPDATE_FREQ == 0:
            if t >= ROLL_WINDOW + DELAY:
                train_end = t - DELAY
                train_start = max(lags, train_end - ROLL_WINDOW)
                if is_dl:
                    model.train()
                    update_X = X_tensor[train_start : train_end]
                    update_y = y_tensor[train_start : train_end]
                    batch_size = 256
                    
                    for i in range(0, len(update_X), batch_size):
                        X_batch = update_X[i : i + batch_size]
                        y_batch = update_y[i : i + batch_size]
                        
                        optimizer.zero_grad()
                        pred_batch = model(X_batch)
                        loss = loss_fn(pred_batch, y_batch)
                        loss.backward()
                        optimizer.step()
                else:
                    ridge_model.fit(X_scaled[train_start : train_end], y_scaled[train_start : train_end])

    val_preds_arr = np.array(predictions).reshape(-1, 1)
    val_preds = scaler_y.inverse_transform(val_preds_arr).flatten()

    if is_hybrid:
        # final_pred_vol = base_vol_test + val_preds
        final_pred_vol = base_vol_test * (np.exp(val_preds))
    else:
        final_pred_vol = val_preds

    return test_dates, actuals, final_pred_vol


def run_annual_oos_evaluation(monthly_experiments, df_full, feature_cols, model_name, scenario_type, dropout_rate=0.0):
    print(f"  -> {model_name} ({scenario_type})...", end=" ")
    all_dates, all_actuals, all_preds = [], [], []
    best_p = BASELINE_PARAMS[model_name] 
    
    for exp in monthly_experiments:
        df_period = df_full.loc[exp['train_start'] : exp['test_end']].copy()
        dates, actuals, preds = run_oos_simulation_for_month(
            df_period=df_period, test_start=exp['test_start'], feature_cols=feature_cols,
            model_name=model_name, params=best_p, scenario_type=scenario_type, dropout_rate=dropout_rate
        )
        all_dates.extend(dates)
        all_actuals.extend(actuals)
        all_preds.extend(preds)
        
    mse = mean_squared_error(all_actuals, all_preds)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(all_actuals, all_preds)
    epsilon = 1e-8
    mape = mean_absolute_percentage_error(np.array(all_actuals) + epsilon, np.array(all_preds) + epsilon)
    r2 = r2_score(all_actuals, all_preds)
    
    print(f"RMSE: {rmse:.4f} | MAE: {mae:.4f} | R2: {r2:.4f}")
    
    df_results = pd.DataFrame({
        'Datetime': all_dates, 
        'Actual_RV': all_actuals, 
        'Predicted_RV': all_preds, 
        'Model_Name': model_name, 
        'Scenario': scenario_type
    })
    
    metrics = {'Model': f"{scenario_type}_{model_name}", 'MSE': mse, 'RMSE': rmse, 'MAE': mae, 'MAPE': mape, 'R-Squared': r2}
    return metrics, df_results
```

```python
if __name__ == "__main__":
    TARGET_YEAR = 2021
    
    EXPERIMENT_DIR = "Dropout_Smoothing"
    USE_SMOOTHING = True
    DROPOUT_RATE = 0.3
    
    ai_models = ["Ridge", "LSTM", "BI-LSTM", "GRU", "BI-GRU"]
    
    os.makedirs(EXPERIMENT_DIR, exist_ok=True)
    checkpoint_file = f"{EXPERIMENT_DIR}/Live_Leaderboard_{TARGET_YEAR}.csv"
    
    completed_models = set()
    if os.path.exists(checkpoint_file):
        df_check = pd.read_csv(checkpoint_file)
        if 'Model' in df_check.columns:
            completed_models = set(df_check['Model'].values)
        print(f"\nSKip: {len(completed_models)}\n")
    
    df_10m_full = pd.concat([df_10m_2020, df_10m_2021]).sort_index()
    if df_10m_full.index.tz is not None: df_10m_full.index = df_10m_full.index.tz_localize(None)
    df_10m_full = df_10m_full[~df_10m_full.index.duplicated(keep='first')]
    df_10m_full = ohlcv_rv(df_10m_full, window=144)
    returns_decimal = df_10m_full['return'] / 100 
    df_10m_full['RV_Return'] = np.sqrt((returns_decimal**2).rolling(window=144).sum())
    # df_10m_full.dropna(subset=['RV_OHLCV'], inplace=True)    

    mfdfa_full = pd.concat([mfdfa_2020, mfdfa_2021]).sort_index()
    if mfdfa_full.index.tz is not None: mfdfa_full.index = mfdfa_full.index.tz_localize(None)
    mfdfa_full = mfdfa_full[~mfdfa_full.index.duplicated(keep='first')]
    
    mfdfa_shifted = mfdfa_full.resample('1D').last().shift(1)
    mfdfa_shifted.index = pd.to_datetime(mfdfa_shifted.index).normalize()
    df_10m_full['temp_date'] = df_10m_full.index.normalize()
    df_10m_full = df_10m_full.merge(mfdfa_shifted, left_on='temp_date', right_index=True, how='left').fillna(method='ffill').dropna()
    df_10m_full.drop(columns=['temp_date'], inplace=True)
    
    # df_10m_full['Future_Vol_24h'] = df_10m_full['RV_OHLCV'].shift(-144)
    df_10m_full['Future_Vol_24h'] = df_10m_full['RV_Return'].shift(-144)
    df_10m_full.dropna(subset=['Future_Vol_24h'], inplace=True)
    
    START_MONTH = 1
    END_MONTH = 12
    monthly_experiments = []
    for month in range(START_MONTH, END_MONTH + 1):
        test_start = pd.Timestamp(f"{TARGET_YEAR}-{month:02d}-01")
        test_end = test_start + relativedelta(months=1) - pd.Timedelta(seconds=1)
        train_start = test_start - relativedelta(months=6)
        train_end = test_start - pd.Timedelta(seconds=1)
        monthly_experiments.append({'target_month': f"{TARGET_YEAR}-{month:02d}", 'train_start': train_start, 'train_end': train_end, 'test_start': test_start, 'test_end': test_end})

    final_leaderboard = []

    print("\n" + "="*50 + "\n[Part 1] Base Statistical Models (4 Models)\n" + "="*50)
    
    daily_returns_pct = df_10m_full['return'].resample('1D').sum().dropna()
    garch11_dates, garch11_preds, garch11_actuals = [], [], []
    
    TRAIN_WINDOW_DAILY = 30 
    
    for exp in monthly_experiments:
        test_period_10m = df_10m_full.loc[exp['test_start'] : exp['test_end']].copy()
        
        test_days = test_period_10m.resample('1D').first().index.normalize()
        
        monthly_garch_preds = []
        monthly_garch_dates = []
        
        for day in test_days:
            train_start_day = day - pd.Timedelta(days=TRAIN_WINDOW_DAILY)
            train_end_day = day - pd.Timedelta(days=1)
            
            train_ret = daily_returns_pct.loc[train_start_day : train_end_day]
            
            if len(train_ret) >= 10: 
                try:
                    am = arch_model(train_ret, vol='GARCH', p=1, q=1)
                    res = am.fit(disp='off')
                    forecast = res.forecast(horizon=1)
                    pred_vol = np.sqrt(forecast.variance.values[-1, 0]) / 100
                except:
                    pred_vol = monthly_garch_preds[-1] if len(monthly_garch_preds) > 0 else 0
            else:
                pred_vol = 0
                
            monthly_garch_preds.append(pred_vol)
            monthly_garch_dates.append(day)
            
        pred_df_daily = pd.DataFrame({'garch_1d': monthly_garch_preds}, index=monthly_garch_dates)
        
        test_period_10m['temp_date'] = test_period_10m.index.normalize()
        test_period_10m = test_period_10m.merge(pred_df_daily, left_on='temp_date', right_index=True, how='left')
        test_period_10m.dropna(subset=['garch_1d', 'Future_Vol_24h'], inplace=True)
        
        garch11_dates.extend(test_period_10m.index)
        garch11_preds.extend(test_period_10m['garch_1d'].values)
        garch11_actuals.extend(test_period_10m['Future_Vol_24h'].values)

    if len(garch11_actuals) > 0:
        mse = mean_squared_error(garch11_actuals, garch11_preds)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(garch11_actuals, garch11_preds)
        mape = mean_absolute_percentage_error(np.array(garch11_actuals) + 1e-8, np.array(garch11_preds) + 1e-8)
        r2 = r2_score(garch11_actuals, garch11_preds)
        final_leaderboard.append({'Model': 'GARCH(1,1)', 'MSE': mse, 'RMSE': rmse, 'MAE': mae, 'MAPE': mape, 'R-Squared': r2})
        print(f"  -> GARCH(1,1) RMSE: {rmse:.4f} | R2: {r2:.4f}")
        
        df_garch11_res = pd.DataFrame({
            'Datetime': garch11_dates, 
            'Actual_RV': garch11_actuals, 
            'Predicted_RV': garch11_preds, 
            'Model_Name': 'GARCH(1,1)', 
            'Scenario': 'Benchmark'
        })
        df_garch11_res.to_csv(f"{EXPERIMENT_DIR}/Predictions_{TARGET_YEAR}_GARCH11.csv", index=False)
    
    garch_scenarios = {"Benchmark": None, "Delta_H": "delta_H", "Delta_Alpha": "delta_alpha"}
    df_hybrid_dict = {} 
    
    for g_name, exog_col in garch_scenarios.items():
        print(f"Extracting GARCH-MIDAS ({g_name}) Data...")
        hybrid_results = []
        month_starts = df_10m_full.resample('MS').first().index
        daily_rv_all = (df_10m_full['RV_Return']**2).resample('1D').last().to_frame('rv').dropna()
        
        if exog_col is None:
            df_x_raw = pd.DataFrame(0, index=daily_rv_all.index, columns=['exog_x'])
        else:
            raw_exog = mfdfa_full[[exog_col]].resample('1D').mean().reindex(daily_rv_all.index).fillna(method='ffill').fillna(0)
            df_x_raw = pd.DataFrame(raw_exog, index=raw_exog.index, columns=[exog_col])
            
        for exp in monthly_experiments:
            train_slice = df_10m_full.loc[exp['train_start'] : exp['train_end']]
            test_slice = df_10m_full.loc[exp['test_start'] : exp['test_end']]
            if exog_col is not None:
                scaler = StandardScaler()
                train_x = df_x_raw.loc[:exp['train_end']] 
                scaler.fit(train_x)
                df_x_full = pd.DataFrame(scaler.transform(df_x_raw), index=df_x_raw.index, columns=[exog_col])
            else:
                df_x_full = df_x_raw.copy()
            try:
                pred_df = garch_midas_tuned(train_slice, test_slice, daily_rv_all, df_x_full, midas_k=BASELINE_PARAMS["GARCH_K"])
                hybrid_results.append(pred_df)
            except Exception as e:
                print(f"GARCH-MIDAS 에러 발생 ({exp['target_month']}): {e}")
            
        df_base_full = pd.concat(hybrid_results).sort_index() if hybrid_results else pd.DataFrame()
        if len(df_base_full) > 0:
            df_base_full = df_base_full[~df_base_full.index.duplicated(keep='first')]
        
        if len(df_base_full) == 0: 
            print(f"[경고] {g_name} 모형 예측 결과가 없어 건너뜁니다.")
            continue
        
        df_base_full['Future_Vol_24h'] = df_10m_full['Future_Vol_24h']

        df_base_full['gm_var'] = df_base_full['gm_vol'] ** 2
        df_base_full['Smoothed_Base_RV'] = np.sqrt(df_base_full['gm_var'].rolling(window=144).mean())
        
        df_eval = df_base_full.dropna(subset=['Future_Vol_24h', 'Smoothed_Base_RV'])
        df_eval = df_eval[df_eval.index.year == TARGET_YEAR]
        
        mse = mean_squared_error(df_eval['Future_Vol_24h'], df_eval['Smoothed_Base_RV'])
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(df_eval['Future_Vol_24h'], df_eval['Smoothed_Base_RV'])
        mape = mean_absolute_percentage_error(df_eval['Future_Vol_24h']+1e-8, df_eval['Smoothed_Base_RV']+1e-8)
        r2 = r2_score(df_eval['Future_Vol_24h'], df_eval['Smoothed_Base_RV'])
        final_leaderboard.append({'Model': f'GARCH-MIDAS Only ({g_name})', 'MSE': mse, 'RMSE': rmse, 'MAE': mae, 'MAPE': mape, 'R-Squared': r2})
        print(f"  -> GARCH-MIDAS ({g_name}) RMSE: {rmse:.4f} | R2: {r2:.4f}")
        df_midas_res = pd.DataFrame({
            'Datetime': df_eval.index,
            'Actual_RV': df_eval['Future_Vol_24h'].values,
            'Predicted_RV': df_eval['Smoothed_Base_RV'].values,
            'Model_Name': f'GARCH-MIDAS_{g_name}',
            'Scenario': 'Benchmark'
        })
        df_midas_res.to_csv(f"{EXPERIMENT_DIR}/Predictions_{TARGET_YEAR}_GARCH_MIDAS_{g_name}.csv", index=False)
        
        df_hybrid_ready = df_10m_full.copy()
        
        df_hybrid_ready = df_hybrid_ready.join(df_base_full[['gm_vol']], how='inner')
        
        df_hybrid_ready['gm_var'] = df_hybrid_ready['gm_vol'] ** 2
        
        if USE_SMOOTHING:
            df_hybrid_ready['Smoothed_Base_RV'] = np.sqrt(df_hybrid_ready['gm_var'].rolling(window=144).mean())
        else:
            df_hybrid_ready['Smoothed_Base_RV'] = df_hybrid_ready['gm_vol']
        
        df_hybrid_ready.dropna(subset=['Smoothed_Base_RV'], inplace=True)
        
        df_hybrid_ready['Base_Vol_24h'] = df_hybrid_ready['Smoothed_Base_RV']
        
        # df_hybrid_ready['Target_Residual'] =df_hybrid_ready['Future_Vol_24h'] - df_hybrid_ready['Base_Vol_24h']
        
        # df_hybrid_ready['Past_Residual'] = df_hybrid_ready['RV_OHLCV'] - df_hybrid_ready['Base_Vol_24h'].shift(144)
        
        df_hybrid_ready['Target_Residual'] =np.log(df_hybrid_ready['Future_Vol_24h']/df_hybrid_ready['Base_Vol_24h'])
        
        df_hybrid_ready['Past_Residual'] = np.log(df_hybrid_ready['RV_Return']/df_hybrid_ready['Base_Vol_24h'].shift(144))
        
        df_hybrid_ready.dropna(subset=['Target_Residual', 'Past_Residual', 'Future_Vol_24h', 'gm_vol'], inplace=True)
        df_hybrid_dict[g_name] = df_hybrid_ready

    print("\n" + "="*50 + "\n[Part 2] Direct AI Models (10 Models)\n" + "="*50)
    
    direct_feature_sets = {
        "No_MFDFA": ['RV_Return', 'return'],
        "With_MFDFA": ['RV_Return', 'return', 'delta_H', 'delta_alpha', 'alpha_0']
    }
    
    for f_name, f_cols in direct_feature_sets.items():
        print(f"\n--- Direct AI [{f_name}] ---")
        for ai_name in ai_models:
            model_id = f"Direct_{f_name}_{ai_name}"
            if model_id in completed_models:
                print(f"  -> {model_id} Already Done.")
                continue
            metrics, df_res = run_annual_oos_evaluation(monthly_experiments, df_10m_full, f_cols, ai_name, f'Direct_{f_name}', dropout_rate=DROPOUT_RATE)
            final_leaderboard.append(metrics)
            df_res.to_csv(f"{EXPERIMENT_DIR}/Predictions_{TARGET_YEAR}_Direct_{f_name}_{ai_name}.csv", index=False)
            pd.DataFrame([metrics]).to_csv(checkpoint_file, mode='a', header=not os.path.exists(checkpoint_file), index=False)

    print("\n" + "="*50 + "\n[Part 3] Hybrid AI Models (30 Models)\n" + "="*50)
    
    for g_name in garch_scenarios.keys():        
        if g_name not in df_hybrid_dict: continue
        
        print(f"\n--- Hybrid Base: GARCH-MIDAS ({g_name}) ---")

        df_hybrid_ready = df_hybrid_dict[g_name] 
        
        hybrid_feature_sets = {
            "No_MFDFA": ['RV_Return', 'return', 'Past_Residual'],
            "With_MFDFA": ['RV_Return', 'return', 'Past_Residual', 'delta_H', 'delta_alpha', 'alpha_0']
        }
        
        for f_name, f_cols in hybrid_feature_sets.items():
            print(f"  [Feature Set: {f_name}]")
            for ai_name in ai_models:
                model_id = f"Hybrid_{g_name}_{f_name}_{ai_name}"
                
                if model_id in completed_models:
                    print(f"  -> {model_id} Already Done.")
                    continue
                metrics, df_res = run_annual_oos_evaluation(monthly_experiments, df_hybrid_ready, f_cols, ai_name, f'Hybrid_{g_name}_{f_name}', dropout_rate=DROPOUT_RATE)
                final_leaderboard.append(metrics)
                df_res.to_csv(f"{EXPERIMENT_DIR}/Preds_Hybrid_{g_name}_{f_name}_{ai_name}.csv", index=False)
                pd.DataFrame([metrics]).to_csv(checkpoint_file, mode='a', header=not os.path.exists(checkpoint_file), index=False)
                
    print("\n" + "="*60 + "\nUltimate 44 Models Leaderboard\n" + "="*60)
    df_leaderboard = pd.DataFrame(final_leaderboard).reset_index(drop=True)
    print(df_leaderboard.to_markdown())
    df_leaderboard.to_csv(f"{EXPERIMENT_DIR}/Ultimate_44_Leaderboard_{TARGET_YEAR}.csv", index=False)
    print("\nAll 44 experiments completed successfully!")
```
